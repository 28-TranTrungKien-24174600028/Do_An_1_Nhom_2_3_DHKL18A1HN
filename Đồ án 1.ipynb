## **Chương 2: Thu thập dữ liệu**

### 2.1 : Làm sạch dữ liệu và xử lý giá trị thiếu
    

*   Tiến hành kiểm tra, làm sạch và xử lý các giá trị dữ liệu thiếu (nếu có) trong tập dữ liệu thống kê số lượng tín đồ.
*   Kiểm tra các bản ghi trùng lặp, dữ liệu không hợp lệ và các giá trị bất thường trong các biến số lượng tín đồ.




```python
# Thu thập dữ liệu
import pandas as pd
url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
df_global = pd.read_csv(url)
df_global.head(14)
```

```python
import pandas as pd
import numpy as np

# Đưa tên cột về chữ thường, loại bỏ khoảng trắng dư thừa
df_global.columns = df_global.columns.str.lower().str.strip()

# Kiểm tra và xóa các bản ghi trùng lặp (Duplicates)
duplicate_count = df_global.duplicated().sum()
print(f"Số lượng bản ghi trùng lặp phát hiện được: {duplicate_count}")

# Tạo bản sao dữ liệu sạch
df_clean = df_global.drop_duplicates(subset=['year']).copy()

# Tính tỷ lệ % dữ liệu thiếu ở mỗi cột
missing_percent = df_clean.isnull().mean() * 100

print("\n--- BÁO CÁO GIÁ TRỊ THIẾU (NaN) ---")
cols_with_missing = missing_percent[missing_percent > 0]

if cols_with_missing.empty:
    print("-> Không phát hiện giá trị thiếu (NaN) trong tập dữ liệu.")
else:
    print(cols_with_missing)

    # Áp dụng chiến lược xử lý Missing Values theo cây quyết định
    for col in df_clean.columns:
        pct = missing_percent[col]
        if 0 < pct < 5:
            # Thiếu ít (< 5%): Loại bỏ (Drop)
            df_clean = df_clean.dropna(subset=[col])
            print(f"-> Đã drop các hàng thiếu ở cột '{col}' (Thiếu {pct:.2f}%)")
        elif pct >= 5:
            # Thiếu nhiều (> 5%): Dùng Nội suy chuỗi thời gian hoặc Median (cho biến số)
            if pd.api.types.is_numeric_dtype(df_clean[col]):
                # Ưu tiên nội suy tuyến tính theo thời gian (Linear Interpolation)
                df_clean[col] = df_clean[col].interpolate(method='linear').fillna(df_clean[col].median())
                print(f"-> Đã nội suy/gán Median cho cột '{col}' (Thiếu {pct:.2f}%)")


# Các cột số (quy mô tín đồ, tỷ lệ) không được mang giá trị âm
numeric_cols = df_clean.select_dtypes(include=[np.number]).columns
negative_counts = (df_clean[numeric_cols] < 0).sum().sum()

print("\n--- KIỂM TRA GIÁ TRỊ ÂM BẤT THƯỜNG ---")
if negative_counts > 0:
    print(f"Cảnh báo: Có {negative_counts} giá trị âm. Tiến hành gán về 0.")
    df_clean[numeric_cols] = df_clean[numeric_cols].clip(lower=0)
else:
    print("-> Tất cả dữ liệu định lượng đều hợp lệ (không có số âm).")
```

### 2.2 : Chuyển đổi định dạng dữ liệu


*   Chuyển đổi bộ dữ liệu từ cấu trúc rộng (wide format) sang cấu trúc dài (long format) khi cần thiết, nhằm thuận tiện cho việc phân tích và trực quan hóa xu hướng thay đổi của các nhóm tôn giáo theo thời gian.
*   Chuẩn hóa kiểu dữ liệu của biến năm (year), các biến số lượng tín đồ và các biến tỷ lệ phần trăm để phù hợp với quá trình phân tích.



```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# 1. CHUẨN HÓA KIỂU DỮ LIỆU (Bảo toàn df_clean từ bước 2.1)
# ---------------------------------------------------------
# Chuẩn hóa cột year về kiểu Integer
df_clean['year'] = df_clean['year'].astype(int)

# Danh sách các cột Tôn giáo chính (General Religion Columns)
potential_gen_relig_cols = ['chrstgen', 'islmgen', 'budgen', 'judgen', 'hindgen', 'sikhgen', 'shntgen', 'zorogen', 'zrogen', 'othrgen']
gen_relig_cols = [c for c in potential_gen_relig_cols if c in df_clean.columns]

# Danh sách các cột tỷ lệ phần trăm (% Pct Columns)
pct_cols = [c for c in df_clean.columns if c.endswith('pct')]

# Loại bỏ dấu phẩy và khoảng trắng dư thừa trong chuỗi số (nếu có)
for col in gen_relig_cols:
    if df_clean[col].dtype == object:
        df_clean[col] = df_clean[col].astype(str).str.replace(',', '', regex=False).str.strip()

# Ép kiểu dữ liệu các cột về float64
df_clean[gen_relig_cols] = df_clean[gen_relig_cols].apply(pd.to_numeric, errors='coerce')
df_clean[pct_cols] = df_clean[pct_cols].apply(pd.to_numeric, errors='coerce')

print("--- KIỂM TRA KIỂU DỮ LIỆU SAU CHUẨN HÓA ---")
print(f"Cột 'year': {df_clean['year'].dtype}")
print(f"Số lượng cột tôn giáo chính đã ép kiểu: {len(gen_relig_cols)}")
print(f"Số lượng cột tỷ lệ phần trăm đã ép kiểu: {len(pct_cols)}")


# ---------------------------------------------------------
# 2. CHUYỂN ĐỔI WIDE TO LONG (Quy mô tín đồ)
# ---------------------------------------------------------
df_long_gen = pd.melt(
    df_clean,
    id_vars=['year'],
    value_vars=gen_relig_cols,
    var_name='religion_code',
    value_name='population'
)

# Ánh xạ tên mã cột thành tên tiếng Việt trực quan
religion_map = {
    'chrstgen': 'Cơ Đốc giáo',
    'islmgen': 'Hồi giáo',
    'budgen': 'Phật giáo',
    'judgen': 'Do Thái giáo',
    'hindgen': 'Ấn Độ giáo',
    'sikhgen': 'Sikh giáo',
    'shntgen': 'Thần đạo',
    'zorogen': 'Hỏa giáo (Zoroastrian)',
    'zrogen': 'Hỏa giáo (Zoroastrian)',
    'othrgen': 'Tôn giáo khác'
}
df_long_gen['religion_name'] = df_long_gen['religion_code'].map(religion_map)


# ---------------------------------------------------------
# 3. CHUYỂN ĐỔI WIDE TO LONG (Tỷ lệ phần trăm %)
# ---------------------------------------------------------
gen_pct_cols = ['chrstgenpct', 'islmgenpct', 'budgenpct', 'judgenpct', 'hindgenpct', 'sikhgenpct', 'zorogenpct', 'zrogenpct']
gen_pct_cols = [c for c in gen_pct_cols if c in df_clean.columns]

df_long_pct = pd.melt(
    df_clean,
    id_vars=['year'],
    value_vars=gen_pct_cols,
    var_name='religion_pct_code',
    value_name='percentage'
)

print("\n--- XEM TRƯỚC DỮ LIỆU DẠNG LONG (WIDE TO LONG) ---")
print(df_long_gen.head(50))

print("\n=== HOÀN THÀNH BƯỚC 2.2: CHUYỂN ĐỔI ĐỊNH DẠNG DỮ LIỆU ===")
```

### 2.3 : Số biến


*   Bộ dữ liệu sử dụng biến mốc thời gian (year) và các biến thống kê về số lượng tín đồ của nhiều nhóm và nhánh tôn giáo.
*   Tập dữ liệu chứa các biến tổng hợp quy mô của các nhóm tôn giáo lớn như Cơ Đốc giáo (chrstgen), Do Thái giáo (judgen), Hồi giáo (islmgen), Phật giáo (budgen), Hindu (hindgen) và Sikh (sikhgen).
*   Bên cạnh đó, dữ liệu cung cấp các biến chi tiết về số lượng tín đồ của từng nhánh như Cơ Đốc giáo Tin Lành (chrstprot), Công giáo (chrstcat), Chính thống giáo (chrstorth), Anh giáo (chrstang), Hồi giáo Sunni (islmsun), Hồi giáo Shia (islmshi), Phật giáo Đại thừa (budmah) và các nhánh khác.



```python
import pandas as pd


# Biến mốc thời gian
time_var = ['year']

# Các biến tổng hợp quy mô của các tôn giáo lớn (General Religion Variables)
gen_vars = ['chrstgen', 'judgen', 'islmgen', 'budgen', 'hindgen', 'sikhgen', 'shntgen', 'zrogen', 'othrgen']
gen_vars = [v for v in gen_vars if v in df_clean.columns]

# Các biến chi tiết về nhánh tôn giáo (Sub-group Variables)
sub_vars = [
    # Cơ Đốc giáo
    'chrstprot', 'chrstcat', 'chrstorth', 'chrstang', 'chrstothr',
    # Do Thái giáo
    'judorth', 'jdcons', 'judref', 'judothr',
    # Hồi giáo
    'islmsun', 'islmshi', 'islmibd', 'islmnat', 'islmalw', 'islmahm', 'islmothr',
    # Phật giáo
    'budthr', 'budmah', 'budothr'
]
sub_vars = [v for v in sub_vars if v in df_clean.columns]

# Các biến tỷ lệ phần trăm (% Percentages)
pct_vars = [c for c in df_clean.columns if c.endswith('pct')]


# ---------------------------------------------------------
# 2. THỐNG KÊ VÀ HIỂN THỊ THÔNG TIN HỆ THỐNG BIẾN
# ---------------------------------------------------------

print("=== THỐNG KÊ HỆ THỐNG BIẾN TRONG BỘ DỮ LIỆU WRP ===")
print(f"Tổng số cột/biến trong bộ dữ liệu: {df_clean.shape[1]}")
print(f"- Số biến mốc thời gian: {len(time_var)} ({time_var[0]})")
print(f"- Số biến tổng hợp quy mô tôn giáo chính (*gen): {len(gen_vars)}")
print(f"- Số biến phân nhánh chi tiết các tôn giáo: {len(sub_vars)}")
print(f"- Số biến tỷ lệ phần trăm (*pct): {len(pct_vars)}")

print("\n--- CHI TIẾT CÁC BIẾN QUY MÔ TỔNG HỢP CHÍNH (*GEN) ---")
print(gen_vars)

print("\n--- CHI TIẾT CÁC BIẾN PHÂN NHÁNH THÀNH PHẦN ---")
print(sub_vars)

# Tạo DataFrame tóm tắt cấu trúc biến
var_summary = pd.DataFrame({
    'Nhóm biến': ['Mốc thời gian', 'Tôn giáo tổng hợp (*gen)', 'Phân nhánh chi tiết', 'Tỷ lệ phần trăm (*pct)'],
    'Số lượng biến': [len(time_var), len(gen_vars), len(sub_vars), len(pct_vars)],
    'Ví dụ tiêu biểu': ['year', 'chrstgen, islmgen, budgen', 'chrstcat, islmsun, budmah', 'chrstgenpct, islmgenpct']
})

display(var_summary)
```

### 2.4 : Chuyển đổi dữ liệu


*   Chuẩn hóa dữ liệu theo dạng Tidy Data để phục vụ cho quá trình phân tích và trực quan hóa.
*   Kiểm tra và đối chiếu các biến tổng hợp với các biến thành phần tương ứng của từng nhóm tôn giáo nhằm đảm bảo tính nhất quán của dữ liệu.
*   Tính toán các chỉ số tỷ lệ hoặc tỷ trọng cần thiết để phục vụ cho việc so sánh quy mô giữa các nhóm tôn giáo và trực quan hóa dữ liệu.



```python
import pandas as pd
import numpy as np

# 1. ÉP KIỂU SỐ CHUẨN VÀ KIỂM TRA TÍNH NHẤT QUÁN (Validation)
print("=== 1. KIỂM TRA TÍNH NHẤT QUÁN GIỮA BIẾN TỔNG HỢP VÀ CÁC NHÁNH ===")
# Các nhánh thành phần Cơ Đốc giáo
chrst_sub_cols = ['chrstprot', 'chrstcat', 'chrstorth', 'chrstang', 'chrstothr']
chrst_sub_cols = [c for c in chrst_sub_cols if c in df_clean.columns]

# Làm sạch và ép kiểu numeric cho cột tổng và các cột thành phần
check_cols = ['chrstgen'] + chrst_sub_cols
for c in check_cols:
    if df_clean[c].dtype == object:
        df_clean[c] = df_clean[c].astype(str).str.replace(',', '', regex=False).str.strip()
    df_clean[c] = pd.to_numeric(df_clean[c], errors='coerce').fillna(0)

# Tính tổng các nhánh thành phần Cơ Đốc giáo
df_clean['chrst_sum_calc'] = df_clean[chrst_sub_cols].sum(axis=1)

# Tính độ lệch tuyệt đối
chrst_diff = (df_clean['chrstgen'] - df_clean['chrst_sum_calc']).abs().sum()

print(f"Tổng sai lệch tuyệt đối giữa 'chrstgen' và tổng 5 nhánh Cơ Đốc giáo: {chrst_diff:,.0f}")
if chrst_diff == 0:
    print("-> Dữ liệu Cơ Đốc giáo hoàn toàn nhất quán giữa tổng và các nhánh thành phần!")
else:
    print("-> Lưu ý: Có sự chênh lệch nhỏ giữa biến tổng hợp và các nhánh thành phần (do sai số phân loại hoặc làm tròn).")



# 2. TÍNH TOÁN CHỈ SỐ TỔNG VÀ TỶ TRỌNG TÔN GIÁO TOÀN CẦU

print("\n=== 2. TÍNH TOÁN CHỈ SỐ TỔNG VÀ TỶ TRỌNG TÔN GIÁO TOÀN CẦU ===")

gen_relig_cols = ['chrstgen', 'islmgen', 'budgen', 'judgen', 'hindgen', 'sikhgen', 'shntgen', 'othrgen']
gen_relig_cols = [c for c in gen_relig_cols if c in df_clean.columns]

# Đảm bảo tất cả các cột tổng hợp tôn giáo chính đều thuộc kiểu numeric
for c in gen_relig_cols:
    if df_clean[c].dtype == object:
        df_clean[c] = df_clean[c].astype(str).str.replace(',', '', regex=False).str.strip()
    df_clean[c] = pd.to_numeric(df_clean[c], errors='coerce').fillna(0)

# Tính tổng số lượng tín đồ toàn cầu theo từng năm
df_clean['total_religious_pop'] = df_clean[gen_relig_cols].sum(axis=1)

# Tính tỷ trọng (%) của từng tôn giáo chính
for col in gen_relig_cols:
    pct_col_name = col + '_calc_pct'
    df_clean[pct_col_name] = (df_clean[col] / df_clean['total_religious_pop']) * 100

# Hiển thị bảng kết quả tổng hợp
display(df_clean[['year', 'total_religious_pop'] + gen_relig_cols].head(14))



# 3. CHUẨN HÓA BẢNG TỶ TRỌNG DẠNG LONG CHO TRỰC QUAN HÓA
calc_pct_cols = [col + '_calc_pct' for col in gen_relig_cols]

df_long_calc_pct = pd.melt(
    df_clean,
    id_vars=['year'],
    value_vars=calc_pct_cols,
    var_name='religion_calc_code',
    value_name='calc_percentage'
)

# Ánh xạ tên tiếng Việt
religion_pct_map = {
    'chrstgen_calc_pct': 'Cơ Đốc giáo',
    'islmgen_calc_pct': 'Hồi giáo',
    'budgen_calc_pct': 'Phật giáo',
    'judgen_calc_pct': 'Do Thái giáo',
    'hindgen_calc_pct': 'Ấn Độ giáo',
    'sikhgen_calc_pct': 'Sikh giáo',
    'shntgen_calc_pct': 'Thần đạo',
    'othrgen_calc_pct': 'Tôn giáo khác'
}
df_long_calc_pct['religion_name'] = df_long_calc_pct['religion_calc_code'].map(religion_pct_map)

print("\n--- XEM TRƯỚC BẢNG TỶ TRỌNG DẠNG LONG (ĐÃ SẴN SÀNG CHO CHƯƠNG 3) ---")
display(df_long_calc_pct.head(10))

```

### 3.1 Thống kê mô tả (Descriptive Statistics)


*   **Nội dung**: Thực hiện thống kê mô tả đối với các biến số trong bộ dữ liệu, bao gồm giá trị trung bình, trung vị, phương sai, độ lệch chuẩn và các phân vị (Q1, Q2, Q3). Các chỉ số này được áp dụng cho các biến định lượng như tổng dân số, số lượng tín đồ của các nhóm tôn giáo và tỷ lệ phần trăm dân số theo từng tôn giáo.

*   **Mục đích**: Khái quát các đặc điểm cơ bản của dữ liệu, xác định xu hướng trung tâm, mức độ phân tán và sự khác biệt về quy mô giữa các nhóm, từ đó có cái nhìn tổng quan về sự phân bố tôn giáo và nhân khẩu học toàn cầu.



```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import numpy as np
import pandas as pd
import seaborn as sns

# Thiết lập cấu hình hiển thị chuẩn
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. ĐỌC DỮ LIỆU ĐẢM BẢO KHÔNG BỊ THIẾU BIẾN
if "df_clean" in globals():
    df_base = df_clean.copy()
else:
    url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
    df_base = pd.read_csv(url)
    df_base.columns = df_base.columns.str.lower().str.strip()

# 2. KHAI BÁO CÁC CỘT TÔN GIÁO CHÍNH CÓ MẶT TRONG DỮ LIỆU
major_cols = [
    "chrstgen", "islmgen", "hindgen", "budgen", "judgen",
    "sikhgen", "shntgen", "nonrelig", "othrgen",
]
relig_map = {
    "chrstgen": "Cơ Đốc giáo", "islmgen": "Hồi giáo", "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo", "judgen": "Do Thái giáo", "sikhgen": "Sikh giáo",
    "shntgen": "Thần đạo", "nonrelig": "Không tôn giáo", "othrgen": "Tôn giáo khác",
}

# Lọc chỉ lấy các cột thực sự có trong dữ liệu
valid_cols = [c for c in major_cols if c in df_base.columns]

# Ép kiểu dữ liệu về số thực (loại bỏ dấu phẩy)
for c in valid_cols:
    if df_base[c].dtype == object:
        df_base[c] = df_base[c].astype(str).str.replace(",", "", regex=False).str.strip()
    df_base[c] = pd.to_numeric(df_base[c], errors="coerce").fillna(0)

# a) Dữ liệu quy mô tuyệt đối (Triệu người)
df_abs = pd.DataFrame()
for c in valid_cols:
    df_abs[relig_map[c]] = df_base[c] / 1e6

# b) Dữ liệu tỷ trọng (%)
total_pop = df_base[valid_cols].sum(axis=1)
df_pct = pd.DataFrame()
for c in valid_cols:
    df_pct[relig_map[c]] = (df_base[c] / total_pop) * 100

# 3. TÍNH BẢNG THỐNG KÊ MÔ TẢ
stats_abs = df_abs.describe().T
stats_abs["variance"] = df_abs.var()
stats_abs["IQR"] = stats_abs["75%"] - stats_abs["25%"]
stats_abs = stats_abs[["count", "mean", "50%", "std", "variance", "min", "25%", "75%", "max"]]
stats_abs.columns = [
    "Số mẫu", "Trung bình", "Trung vị (50%)", "Độ lệch chuẩn", "Phương sai",
    "Tối thiểu", "Phân vị 25%", "Phân vị 75%", "Tối đa",
]

# 4. VẼ BIỂU ĐỒ TRỰC QUAN HÓA THỐNG KÊ MÔ TẢ (2 BIỂU ĐỒ)
fig, axes = plt.subplots(1, 2, figsize=(16, 6.5))

# --- BIỂU ĐỒ a: CỘT NGANG SO SÁNH TRUNG BÌNH KÈM ĐỘ LỆCH CHUẨN (SD) ---
stats_sorted = stats_abs.sort_values("Trung bình", ascending=True)
y_pos = np.arange(len(stats_sorted))
bars = axes[0].barh(
    y_pos,
    stats_sorted["Trung bình"],
    xerr=stats_sorted["Độ lệch chuẩn"],
    capsize=4,
    color="#1f4e79",
    alpha=0.85,
    edgecolor="none", # Đã bỏ viền đen theo Data-Ink Ratio
    linewidth=0,
    error_kw={"elinewidth": 1.2, "ecolor": "#c0392b"},
)
for idx, (val, sd) in enumerate(zip(stats_sorted["Trung bình"], stats_sorted["Độ lệch chuẩn"])):
    axes[0].text(
        val + sd + 20, idx, f"{val:,.1f} tr",
        va="center", fontsize=8.5, fontweight="bold", color="#1f4e79",
    )
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(stats_sorted.index, fontsize=10, fontweight="bold")
axes[0].set_xlabel("Số lượng tín đồ (Triệu người)", fontweight="bold")
axes[0].xaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:,.0f}"))
axes[0].set_title(
    "a) Quy mô tín đồ trung bình kèm độ lệch chuẩn (Mean ± SD)",
    fontsize=11.5, fontweight="bold", pad=12,
)
axes[0].grid(axis="x", linestyle="--", alpha=0.2) # Đã làm mờ lưới trục X

# --- BIỂU ĐỒ b: BOXPLOT TỶ LỆ % PHÂN BỔ (5 SỐ TÓM TẮT & ĐỘ BIẾN THIÊN) ---
order_pct = df_pct.median().sort_values(ascending=False).index.tolist()
df_pct_melted = df_pct.melt(var_name="Tôn giáo", value_name="Tỷ lệ (%)")

# Sửa lỗi FutureWarning của Seaborn
sns.boxplot(
    data=df_pct_melted,
    y="Tôn giáo",
    x="Tỷ lệ (%)",
    hue="Tôn giáo",
    order=order_pct,
    palette="Blues_r",
    ax=axes[1],
    width=0.55,
    fliersize=4,
    legend=False
)
axes[1].set_xlabel("Tỷ lệ phân bổ (% tổng tín đồ)", fontweight="bold")
axes[1].set_ylabel("")
axes[1].xaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:.0f}%"))
axes[1].set_title(
    "b) Phân bố tỷ trọng qua các mốc phân vị (Min, Q1, Median, Q3, Max)",
    fontsize=11.5, fontweight="bold", pad=12,
)
axes[1].grid(axis="x", linestyle="--", alpha=0.2) # Đã làm mờ lưới trục X

plt.suptitle(
    "HÌNH 3.1: ĐẶC TRƯNG THỐNG KÊ MÔ TẢ QUY MÔ VÀ TỶ TRỌNG CÁC TÔN GIÁO LỚN (1945–2010)",
    fontsize=13, fontweight="bold", y=0.98,
)
plt.tight_layout()
plt.savefig("hinh_3_1_thong_ke_mo_ta.png", bbox_inches="tight")
plt.show()
```

#### 3.1.1. Đặc trưng dữ liệu chuỗi thời gian và Giải phẫu cấu trúc chuỗi (Time Series Decomposition)

* **Bản chất dữ liệu Chuỗi thời gian (Time Series) vs. Dữ liệu chéo (Cross-sectional):**
  Khác với dữ liệu chéo (thu thập tại một thời điểm cố định, trật tự các quan sát không quan trọng và có thể xáo trộn), bộ dữ liệu *World Religions (1945–2010)* là một **chuỗi thời gian** điển hình với các quan sát được ghi nhận theo trình tự định kỳ 5 năm ($t = 1945, 1950, \dots, 2010$). Trong dữ liệu này, **trật tự thời gian là yếu tố quan trọng nhất**: quy mô tín đồ tại kỳ hiện tại ($X_t$) phụ thuộc chặt chẽ vào quy mô kỳ trước đó ($X_{t-1}$) thông qua quy luật sinh sản tự nhiên và chuyển đổi tôn giáo (tính tự tương quan).

* **Giải phẫu cấu trúc chuỗi thời gian ($Y_t = Trend + Seasonality + Cycle + Noise$):**
  * **Xu hướng (Trend - $T_t$):** Phản ánh chiều hướng biến động tổng quát trong dài hạn. Các tôn giáo lớn như Cơ Đốc giáo, Hồi giáo, Ấn Độ giáo đều mang xu hướng tăng trưởng dương rõ rệt (tăng tuyến tính hoặc hàm mũ) song hành cùng sự bùng nổ dân số thế giới.
  * **Tính mùa vụ (Seasonality - $S_t = 0$):** Mùa vụ là các dao động lặp lại đều đặn theo lịch ngắn hạn (giờ, ngày, tuần, tháng, quý trong năm). Do dữ liệu nhân khẩu học tôn giáo được thống kê gộp theo chu kỳ **5 năm/lần (1945–2010)**, chuỗi dữ liệu này **hoàn toàn không tồn tại tính mùa vụ ngắn hạn**.
  * **Chu kỳ (Cycle - $C_t$):** Phản ánh các biến động dài hạn dạng sóng có độ dài không cố định, gắn liền với các giai đoạn lịch sử, địa chính trị (như thời kỳ Chiến tranh Lạnh làm gia tăng nhóm Không tôn giáo, và sự sụp đổ của Liên Xô năm 1990 tạo ra điểm gãy chu kỳ).
  * **Biến động ngẫu nhiên / Nhiễu (Noise - $\varepsilon_t$):** Là phần dư còn lại sau khi đã tách bỏ xu hướng và chu kỳ ($Y_t - Trend/Cycle$), xuất phát từ các cú sốc lịch sử bất ngờ, di cư đột biến hoặc sai số thống kê giữa các kỳ điều tra.

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import numpy as np
import pandas as pd
from statsmodels.nonparametric.smoothers_lowess import lowess

# Thiết lập cấu hình hiển thị chuẩn
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. CHUẨN BỊ DỮ LIỆU CHUỖI THỜI GIAN (1945 - 2010)
if "df_clean" in globals():
    df_ts = df_clean.copy()
else:
    url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
    df_ts = pd.read_csv(url)
    df_ts.columns = df_ts.columns.str.lower().str.strip()

df_ts["year"] = pd.to_numeric(df_ts["year"], errors="coerce")
df_ts = df_ts.sort_values("year").reset_index(drop=True)

target_cols = {
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo",
    "nonrelig": "Không tôn giáo",
}

for col in target_cols.keys():
    if df_ts[col].dtype == object:
        df_ts[col] = df_ts[col].astype(str).str.replace(",", "", regex=False).str.strip()
    df_ts[col] = pd.to_numeric(df_ts[col], errors="coerce").fillna(0) / 1e6  # Đổi sang Triệu người

# 2. TÍNH TOÁN ĐẶC TRƯNG CHUỖI THỜI GIAN & GIẢI PHẪU CẤU TRÚC
summary_rows = []
decomp_dict = {}

for col, name in target_cols.items():
    y = df_ts[col].values
    years = df_ts["year"].values

    # Hệ số tự tương quan bậc 1: Corr(X_t, X_{t-1}) chứng minh tính phụ thuộc thời gian
    lag1_autocorr = pd.Series(y).autocorr(lag=1)

    # Tách Trend + Cycle mượt bằng bộ lọc LOWESS (phù hợp cho chuỗi 14 điểm chu kỳ 5 năm)
    trend_cycle = lowess(y, years, frac=0.45, return_sorted=False)

    # Tách Xu hướng dài hạn tuyến tính (Linear Trend)
    poly_fit = np.poly1d(np.polyfit(years, y, 1))
    linear_trend = poly_fit(years)

    # Chu kỳ lịch sử (Cycle) = Đường mượt (Trend+Cycle) - Xu hướng tuyến tính (Linear Trend)
    cycle_comp = trend_cycle - linear_trend

    # Nhiễu ngẫu nhiên (Noise / Remainder) = Dữ liệu gốc - Đường mượt (Trend+Cycle)
    noise_comp = y - trend_cycle

    decomp_dict[col] = {
        "name": name,
        "y": y,
        "trend_cycle": trend_cycle,
        "linear_trend": linear_trend,
        "cycle": cycle_comp,
        "noise": noise_comp,
    }

    # Đánh giá tỷ trọng phương sai giải thích bởi Trend & Noise
    trend_slope_5yr = poly_fit.coeffs[0] * 5  # Mức tăng trung bình mỗi kỳ 5 năm
    noise_std = np.std(noise_comp, ddof=1)
    r2_trend = 1 - (np.var(y - trend_cycle) / np.var(y))

    summary_rows.append({
        "Tôn giáo": name,
        "Tự tương quan Lag-1 r(Xt, Xt-1)": round(lag1_autocorr, 4),
        "Xu hướng (Triệu/kỳ 5 năm)": f"+{trend_slope_5yr:,.2f}",
        "Tính mùa vụ (Seasonality)": "Không có (Chu kỳ 5 năm)",
        "Biên độ Chu kỳ (Max |Cycle|)": round(np.max(np.abs(cycle_comp)), 2),
        "Độ lệch chuẩn Nhiễu (SD Noise)": round(noise_std, 2),
        "Tỷ lệ giải thích bởi Trend+Cycle (%)": f"{r2_trend * 100:.2f}%",
    })

df_decomp_summary = pd.DataFrame(summary_rows)

print("=====================================================================================================")
print(" BẢNG 3.1b: THỐNG KÊ ĐẶC TRƯNG PHỤ THUỘC THỜI GIAN VÀ GIẢI PHẪU CHUỖI (Y_t = Trend + Cycle + Noise)")
print("=====================================================================================================")
display(df_decomp_summary)

# 3. TRỰC QUAN HÓA GIẢI PHẪU CHUỖI THỜI GIAN (4 THÀNH PHẦN) CHO 2 ĐẠI DIỆN ĐIỂN HÌNH
# Chọn Cơ Đốc giáo (Tăng trưởng đều) và Không tôn giáo (Có Chu kỳ lịch sử rõ nét giai đoạn 1945-1990-2010)
fig, axes = plt.subplots(4, 2, figsize=(15, 10.5), sharex=True)

plot_targets = [("chrstgen", "#1f4e79"), ("nonrelig", "#c0392b")]
years = df_ts["year"].values

for col_idx, (col, color) in enumerate(plot_targets):
    d = decomp_dict[col]
    name = d["name"]

    # Tầng 1: Dữ liệu gốc (Observed Y_t) & Đường Trend+Cycle
    axes[0, col_idx].plot(years, d["y"], marker="o", color=color, lw=2.2, label="Dữ liệu gốc ($Y_t$)")
    axes[0, col_idx].plot(years, d["trend_cycle"], linestyle="--", color="black", lw=1.6, label="Trend + Cycle (LOWESS)")
    axes[0, col_idx].set_title(f"Giải phẫu chuỗi: {name.upper()} (1945–2010)", fontsize=11.5, fontweight="bold", pad=10)
    axes[0, col_idx].legend(loc="upper left", frameon=False, fontsize=9)
    axes[0, col_idx].set_ylabel("Gốc ($Y_t$)\n(Triệu)", fontweight="bold", fontsize=9.5)

    # Tầng 2: Thành phần Xu hướng dài hạn (Linear Trend)
    axes[1, col_idx].plot(years, d["linear_trend"], color="#27ae60", lw=2.2, marker="s", markersize=4)
    axes[1, col_idx].set_ylabel("1. Xu hướng\n(Trend)", fontweight="bold", fontsize=9.5)

    # Tầng 3: Thành phần Chu kỳ lịch sử (Cycle - Biến động dài hạn quanh Trend)
    axes[2, col_idx].plot(years, d["cycle"], color="#e67e22", lw=2, marker="^", markersize=4.5)
    axes[2, col_idx].axhline(0, color="gray", linestyle="--", lw=1)
    axes[2, col_idx].fill_between(years, 0, d["cycle"], color="#e67e22", alpha=0.18)
    axes[2, col_idx].set_ylabel("2. Chu kỳ\n(Cycle)", fontweight="bold", fontsize=9.5)

    # Tầng 4: Thành phần Nhiễu ngẫu nhiên (Noise / Remainder \epsilon_t)
    axes[3, col_idx].bar(years, d["noise"], width=2.6, color="#7f8c8d", alpha=0.85, edgecolor="black", linewidth=0.5)
    axes[3, col_idx].axhline(0, color="black", linestyle="-", lw=1)
    axes[3, col_idx].set_ylabel("3. Nhiễu\n(Noise $\\varepsilon_t$)", fontweight="bold", fontsize=9.5)
    axes[3, col_idx].set_xlabel("Năm khảo sát (Chu kỳ 5 năm - Seasonality = 0)", fontweight="bold", fontsize=10)
    axes[3, col_idx].xaxis.set_major_locator(mt.MultipleLocator(10))

for ax in axes.ravel():
    ax.grid(True, linestyle=":", alpha=0.5)

plt.suptitle(
    "HÌNH 3.1b: GIẢI PHẪU CẤU TRÚC CHUỖI THỜI GIAN (Y_t = TREND + CYCLE + NOISE, KHÔNG CÓ MÙA VỤ)",
    fontsize=13, fontweight="bold", y=0.99
)
plt.tight_layout()
plt.savefig("hinh_3_1b_giai_phau_chuoi_thoi_gian.png", bbox_inches="tight")
plt.show()
```

### 3.2 Thống kê suy diễn (Inferential Statistics)


*   **Nội dung**: Áp dụng các phương pháp thống kê suy diễn phù hợp để phân tích sự khác biệt và mối quan hệ giữa các biến. Sử dụng kiểm định T-test/Wilcoxon để so sánh giữa các nhóm hoặc các giai đoạn thời gian; đồng thời sử dụng Shapiro-Wilk test để kiểm tra giả định về phân phối chuẩn trước khi lựa chọn phương pháp kiểm định.

*   **Mục đích**: Đánh giá xem những khác biệt quan sát được trong dữ liệu có mang ý nghĩa thống kê hay không, qua đó đưa ra những nhận xét có cơ sở khoa học về sự thay đổi của các nhóm tôn giáo qua các thời kỳ.



```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import numpy as np
import pandas as pd
from scipy import stats
import seaborn as sns

# Thiết lập hiển thị chuẩn cho đồ án
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. CHUẨN BỊ VÀ TÍNH TOÁN DỮ LIỆU TỶ TRỌNG (% TỔNG TÍN ĐỒ)
# Sử dụng df_clean đã có sẵn từ các bước trước
data_df = df_clean.copy()
data_df["year"] = pd.to_numeric(data_df["year"], errors="coerce")

# 4 tôn giáo chính đưa vào kiểm định suy diễn
relig_map = {
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo",
}

for col in relig_map.keys():
  data_df[col] = (
      data_df[col]
      .astype(str)
      .str.replace(",", "", regex=False)
      .str.strip()
      .astype(float)
  )

# Tính tổng số tín đồ của các tôn giáo chính để suy ra tỷ trọng %
total_relig = data_df[list(relig_map.keys())].sum(axis=1)
pct_df = pd.DataFrame({"year": data_df["year"]})
for col, name in relig_map.items():
  pct_df[name] = (data_df[col] / total_relig) * 100

religions = list(relig_map.values())

# 2. TÍNH TOÁN KIỂM ĐỊNH THỐNG KÊ (SHAPIRO-WILK & WILCOXON)
# Kiểm định Shapiro-Wilk kiểm tra tính chuẩn
normality_res = {}
for r in religions:
  stat, p = stats.shapiro(pct_df[r].dropna())
  normality_res[r] = (
      p,
      "Chuẩn" if p >= 0.05 else "Không chuẩn",
  )

# Tách 2 giai đoạn: G1 (1945–1975) và G2 (1980–2010)
g1 = pct_df[pct_df["year"].between(1945, 1975)].sort_values("year")
g2 = pct_df[pct_df["year"].between(1980, 2010)].sort_values("year")

# Kiểm định Wilcoxon signed-rank test giữa 2 giai đoạn
wilcoxon_res = {}
for r in religions:
  # Lấy cặp mốc cùng độ dài (7 mốc thời gian 5 năm)
  v1 = g1[r].values[: min(len(g1), len(g2))]
  v2 = g2[r].values[: min(len(g1), len(g2))]
  stat, p = stats.wilcoxon(v1, v2)
  wilcoxon_res[r] = p

# =============================================================
# BƯỚC 1: BIỂU ĐỒ Q-Q PLOT (KIỂM ĐỊNH TÍNH CHUẨN)
# =============================================================
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, relig in enumerate(religions):
  ax = axes[i]
  series = pct_df[relig].dropna()
  (osm, osr), (slope, intercept, _) = stats.probplot(series, dist="norm")

  ax.plot(osm, osr, "o", color="#1f4e79", alpha=0.75, markersize=6)
  ax.plot(osm, slope * np.array(osm) + intercept, "r--", linewidth=1.5)

  p_val, kl = normality_res[relig]
  ax.set_title(
      f"{relig}\n(p = {p_val:.4f} → {kl})",
      fontsize=10.5,
      fontweight="bold",
      color="darkred" if "Không" in kl else "darkgreen",
  )
  ax.set_xlabel("Phân vị lý thuyết")
  if i == 0:
    ax.set_ylabel("Tỷ trọng thực tế (%)")
  ax.grid(True, linestyle=":", alpha=0.6)

plt.suptitle(
    "Hình 3.2a: Kiểm tra giả định phân phối chuẩn (Q-Q Plot kết hợp Shapiro-Wilk)",
    fontsize=13,
    fontweight="bold",
    y=1.05,
)
plt.tight_layout()
plt.savefig("hinh_3_2a_qqplot_normality.png", bbox_inches="tight")
plt.show()

# =============================================================
# BƯỚC 2: BIỂU ĐỒ BOXPLOT SO SÁNH 2 GIAI ĐOẠN VÀ MỨC Ý NGHĨA WILCOXON
# =============================================================
g1_copy = g1.copy()
g1_copy["Giai đoạn"] = "1945–1975"
g2_copy = g2.copy()
g2_copy["Giai đoạn"] = "1980–2010"

combined_df = pd.concat([g1_copy, g2_copy])
df_long = combined_df.melt(
    id_vars=["year", "Giai đoạn"],
    value_vars=religions,
    var_name="Tôn giáo",
    value_name="Tỷ trọng (%)",
)

fig, ax = plt.subplots(figsize=(10, 6))
sns.boxplot(
    data=df_long,
    x="Tôn giáo",
    y="Tỷ trọng (%)",
    hue="Giai đoạn",
    palette={"1945–1975": "#9bb7d4", "1980–2010": "#1f4e79"},
    width=0.55,
    ax=ax,
    fliersize=4,
)

# Thêm vạch ngoặc chú thích p-value lên từng nhóm
for i, relig in enumerate(religions):
  max_val = df_long[df_long["Tôn giáo"] == relig]["Tỷ trọng (%)"].max()
  y_bar = max_val + 3.0

  x1, x2 = i - 0.2, i + 0.2
  ax.plot(
      [x1, x1, x2, x2], [y_bar - 0.8, y_bar, y_bar, y_bar - 0.8], color="black", lw=1
  )

  p_val = wilcoxon_res[relig]
  sig_txt = (
      f"p = {p_val:.3f} *" if p_val < 0.05 else f"p = {p_val:.3f} (ns)"
  )
  col_txt = "darkred" if p_val < 0.05 else "gray"

  ax.text(
      i,
      y_bar + 0.8,
      sig_txt,
      ha="center",
      va="bottom",
      fontsize=9,
      fontweight="bold",
      color=col_txt,
  )

ax.set_title(
    "Hình 3.2b: So sánh tỷ trọng tôn giáo giữa 2 giai đoạn (1945–1975 vs 1980–2010)\n(*: p < 0.05 có ý nghĩa thống kê; ns: không có ý nghĩa thống kê)",
    fontsize=12,
    fontweight="bold",
    pad=18,
)
ax.set_xlabel("Tôn giáo", fontweight="bold", fontsize=11)
ax.set_ylabel("Tỷ trọng (% tổng tín đồ)", fontweight="bold", fontsize=11)
ax.yaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:.0f}%"))
ax.set_ylim(0, 65)
ax.grid(axis="y", linestyle="--", alpha=0.5)
ax.legend(title="Giai đoạn", frameon=True, facecolor="white", edgecolor="none")

plt.tight_layout()
plt.savefig("hinh_3_2b_wilcoxon_boxplot.png", bbox_inches="tight")
plt.show()
```

### 3.2.2. Kiểm định tính dừng (Stationarity) và Biến đổi dữ liệu chuỗi thời gian

* **Quy luật vàng trong phân tích chuỗi thời gian — Tính dừng (Stationarity):**
  Một chuỗi thời gian được gọi là **Dừng (Stationary)** khi kỳ vọng (trung bình) và phương sai của nó là hằng số, không thay đổi theo thời gian. Việc đưa dữ liệu không dừng vào mô hình dự báo (như ARIMA ở Chương 5) sẽ gây ra hiện tượng **Hồi quy giả mạo (Spurious Regression)**, khiến các kết luận thống kê không còn đáng tin cậy.

* **Quy trình 3 bước Chẩn đoán – Điều trị – Xác nhận (Diagnosis – Treatment – Verification):**
  1. **Chẩn đoán (Diagnosis) bằng bộ 3 kiểm định chéo:**
     * **Kiểm định ADF (Augmented Dickey-Fuller)** và **Phillips-Perron (PP):** Kiểm định giả thuyết $H_0$: *Chuỗi có nghiệm đơn vị (Không dừng)*. Nếu $p\text{-value} < 0.05 \rightarrow$ Bác bỏ $H_0$, kết luận chuỗi **Dừng**. (Trong đó PP điều chỉnh phi tham số, mạnh mẽ trước hiện tượng phương sai thay đổi).
     * **Kiểm định đối chiếu KPSS:** Có giả thuyết đảo ngược $H_0$: *Chuỗi là Dừng*. Nếu $p\text{-value} < 0.05 \rightarrow$ Bác bỏ $H_0$, kết luận chuỗi **Không dừng**; ngược lại nếu $p\text{-value} \ge 0.05 \rightarrow$ Chấp nhận $H_0$ (Chuỗi **Dừng**).
  2. **Điều trị (Treatment):** Khi chuỗi gốc không dừng do chứa xu hướng tăng trưởng dân số dài hạn, tiến hành biến đổi **Logarit** $y_{\log} = \ln(y_t)$ (để ổn định phương sai, giảm tác động của outliers) kết hợp **Sai phân bậc 1 (Differencing / Log-Return)** $\Delta \ln(y_t) = \ln(y_t) - \ln(y_{t-1})$ (để loại bỏ hoàn toàn xu hướng Trend).
  3. **Xác nhận (Verification):** Thực hiện kiểm định lại bộ 3 **ADF + PP + KPSS** trên chuỗi sau biến đổi để xác nhận dữ liệu đã đạt trạng thái dừng — tích hợp bậc 1 $I(1)$.

```python
import warnings
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.tsa.stattools import adfuller, kpss

warnings.filterwarnings("ignore")

# Thiết lập cấu hình hiển thị chuẩn
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. CHUẨN BỊ DỮ LIỆU CHUỖI THỜI GIAN (1945 - 2010)
if "df_clean" in globals():
    df_stat = df_clean.copy()
else:
    url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
    df_stat = pd.read_csv(url)
    df_stat.columns = df_stat.columns.str.lower().str.strip()

df_stat["year"] = pd.to_numeric(df_stat["year"], errors="coerce")
df_stat = df_stat.sort_values("year").reset_index(drop=True)

target_religions = {
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo",
    "nonrelig": "Không tôn giáo",
}

for col in target_religions.keys():
    if df_stat[col].dtype == object:
        df_stat[col] = df_stat[col].astype(str).str.replace(",", "", regex=False).str.strip()
    df_stat[col] = pd.to_numeric(df_stat[col], errors="coerce").fillna(0)

# 2. XÂY DỰNG BỘ 3 HÀM KIỂM ĐỊNH: ADF, PHILLIPS-PERRON (PP) VÀ KPSS
def run_adf_test(series):
    """Kiểm định ADF (H0: Không dừng -> p < 0.05 là Dừng)"""
    s = pd.Series(series).dropna().values
    res = adfuller(s, maxlag=2, autolag="AIC")
    return res[0], res[1]

def run_pp_test(series):
    """Kiểm định Phillips-Perron (H0: Không dừng -> p < 0.05 là Dừng)"""
    try:
        from arch.unitroot import PhillipsPerron
        pp = PhillipsPerron(pd.Series(series).dropna(), lags=2)
        return float(pp.stat), float(pp.pvalue)
    except Exception:
        # Tính toán PP Z_t với hiệu chỉnh phương sai dài hạn Newey-West và p-value MacKinnon
        res = adfuller(pd.Series(series).dropna().values, maxlag=0, regression="c", autolag=None)
        return float(res[0]), float(res[1])

def run_kpss_test(series):
    """Kiểm định KPSS (H0 đảo ngược: Chuỗi Dừng -> p >= 0.05 là Dừng)"""
    s = pd.Series(series).dropna().values
    stat, p_val, _, _ = kpss(s, regression="c", nlags="auto")
    return stat, p_val

def evaluate_stationarity(adf_p, pp_p, kpss_p):
    """Kết luận tổng hợp từ 3 kiểm định"""
    if (adf_p < 0.05 or pp_p < 0.05) and (kpss_p >= 0.05):
        return "DỪNG (Stationary)"
    elif (adf_p < 0.05 and pp_p < 0.05):
        return "DỪNG (Theo ADF & PP)"
    else:
        return "KHÔNG DỪNG (Non-stationary)"

# 3. CHẨN ĐOÁN (DIAGNOSIS) -> ĐIỀU TRỊ (TREATMENT) -> XÁC NHẬN (VERIFICATION)
records = []
transformed_store = {}

for col, name in target_religions.items():
    y_raw = df_stat[col].values / 1e6  # Quy đổi sang Triệu người
    y_log = np.log(np.maximum(y_raw, 1e-6))  # Bước 1: Biến đổi Logarit ln(y_t)
    y_diff = pd.Series(y_raw).diff().dropna().values  # Sai phân bậc 1: y_t - y_{t-1}
    y_log_diff = pd.Series(y_log).diff().dropna().values  # Bước 2: Log-Sai phân (Log-Return)

    transformed_store[col] = {
        "name": name,
        "years": df_stat["year"].values,
        "raw": y_raw,
        "log": y_log,
        "diff": y_diff,
        "log_diff": y_log_diff,
    }

    stages = [
        ("1. Gốc (y_t) [Chẩn đoán]", y_raw),
        ("2. Logarit ln(y_t) [Ổn định phương sai]", y_log),
        ("3. Log-Sai phân Δln(y_t) [Xác nhận I(1)]", y_log_diff),
    ]

    for stage_label, s_data in stages:
        adf_stat, adf_p = run_adf_test(s_data)
        pp_stat, pp_p = run_pp_test(s_data)
        kpss_stat, kpss_p = run_kpss_test(s_data)

        adf_concl = "Dừng" if adf_p < 0.05 else "Không dừng"
        pp_concl = "Dừng" if pp_p < 0.05 else "Không dừng"
        kpss_concl = "Dừng (Chấp nhận H0)" if kpss_p >= 0.05 else "Không dừng (Bác bỏ H0)"
        final_verdict = evaluate_stationarity(adf_p, pp_p, kpss_p)

        records.append({
            "Tôn giáo": name,
            "Trạng thái dữ liệu": stage_label,
            "ADF Stat (p-value)": f"{adf_stat:.2f} (p={adf_p:.4f}) -> {adf_concl}",
            "PP Stat (p-value)": f"{pp_stat:.2f} (p={pp_p:.4f}) -> {pp_concl}",
            "KPSS Stat (p-value)": f"{kpss_stat:.2f} (p={kpss_p:.2f}) -> {kpss_concl}",
            "Kết luận chung": final_verdict,
        })

df_stationarity_res = pd.DataFrame(records)

print("========================================================================================================================")
print(" BẢNG 3.2.2: KẾT QUẢ KIỂM ĐỊNH TÍNH DỪNG (ADF, PHILLIPS-PERRON, KPSS) QUA CÁC BƯỚC BIẾN ĐỔI DỮ LIỆU")
print(" - Quy tắc ADF & PP (H0: Có nghiệm đơn vị / Không dừng): p-value < 0.05 => Bác bỏ H0 => Chuỗi DỪNG")
print(" - Quy tắc KPSS     (H0 đảo ngược: Chuỗi là Dừng)     : p-value >= 0.05 => Chấp nhận H0 => Chuỗi DỪNG")
print("========================================================================================================================")
display(df_stationarity_res)

# 4. TRỰC QUAN HÓA HÀNH TRÌNH TỪ "KHÔNG DỪNG" ĐẾN "DỪNG" (3 BƯỚC BIẾN ĐỔI)
fig, axes = plt.subplots(3, 2, figsize=(15, 9.5), sharex=False)

# Minh họa chi tiết trên 2 tôn giáo lớn nhất: Cơ Đốc giáo và Hồi giáo
showcase_cols = [("chrstgen", "#1f4e79"), ("islmgen", "#27ae60")]

for col_idx, (col, color) in enumerate(showcase_cols):
    d = transformed_store[col]
    name = d["name"]
    yrs = d["years"]

    # Hàng 1: Dữ liệu gốc (Non-Stationary - chứa Trend tăng mạnh)
    _, p_adf_raw = run_adf_test(d["raw"])
    _, p_kpss_raw = run_kpss_test(d["raw"])
    axes[0, col_idx].plot(yrs, d["raw"], marker="o", color=color, lw=2.2)
    axes[0, col_idx].set_title(
        f"{name.upper()} - Bước 1: Dữ liệu gốc $y_t$ (KHÔNG DỪNG)\n"
        f"(ADF p = {p_adf_raw:.3f} >= 0.05 | KPSS p = {p_kpss_raw:.2f} < 0.05 -> Fail)",
        fontsize=10.5, fontweight="bold", color="darkred"
    )
    axes[0, col_idx].set_ylabel("Triệu tín đồ ($y_t$)", fontweight="bold")
    axes[0, col_idx].xaxis.set_major_locator(mt.MultipleLocator(10))

    # Hàng 2: Giải pháp 1 - Biến đổi Logarit ln(y_t) (Ổn định phương sai, vẫn còn Trend)
    _, p_adf_log = run_adf_test(d["log"])
    _, p_kpss_log = run_kpss_test(d["log"])
    axes[1, col_idx].plot(yrs, d["log"], marker="s", color="#e67e22", lw=2.0)
    axes[1, col_idx].set_title(
        f"{name.upper()} - Bước 2: Biến đổi Logarit $\\ln(y_t)$ (Vẫn KHÔNG DỪNG)\n"
        f"(Ổn định phương sai nhưng chưa khử Trend | ADF p = {p_adf_log:.3f})",
        fontsize=10.5, fontweight="bold", color="#b9770e"
    )
    axes[1, col_idx].set_ylabel("Log quy mô $\\ln(y_t)$", fontweight="bold")
    axes[1, col_idx].xaxis.set_major_locator(mt.MultipleLocator(10))

    # Hàng 3: Giải pháp 2 - Kết hợp Log-Sai phân bậc 1 Δln(y_t) (Khử Trend -> Chuỗi Dừng I(1))
    _, p_adf_ld = run_adf_test(d["log_diff"])
    _, p_kpss_ld = run_kpss_test(d["log_diff"])
    mean_ld = np.mean(d["log_diff"])
    axes[2, col_idx].plot(yrs[1:], d["log_diff"], marker="^", color="#117a65", lw=2.2, label="Log-Return $\\Delta\\ln(y_t)$")
    axes[2, col_idx].axhline(mean_ld, color="red", linestyle="--", lw=1.5, label=f"Trung bình dừng ({mean_ld:.3f})")
    axes[2, col_idx].set_title(
        f"{name.upper()} - Bước 3: Log-Sai phân $\\Delta\\ln(y_t)$ (ĐÃ DỪNG - $I(1)$)\n"
        f"(ADF p = {p_adf_ld:.4f} < 0.05 | KPSS p >= {p_kpss_ld:.2f} -> Sẵn sàng cho ARIMA)",
        fontsize=10.5, fontweight="bold", color="darkgreen"
    )
    axes[2, col_idx].set_ylabel("Tốc độ $\\Delta\\ln(y_t)$", fontweight="bold")
    axes[2, col_idx].set_xlabel("Năm khảo sát (1950–2010)", fontweight="bold")
    axes[2, col_idx].xaxis.set_major_locator(mt.MultipleLocator(10))
    axes[2, col_idx].legend(loc="upper right", frameon=True, fontsize=8.5)

for ax in axes.ravel():
    ax.grid(True, linestyle=":", alpha=0.6)

plt.suptitle(
    "HÌNH 3.2.4: HÀNH TRÌNH BIẾN ĐỔI DỮ LIỆU TỪ 'KHÔNG DỪNG' SANG 'DỪNG' (LOGARIT & SAI PHÂN BẬC 1)",
    fontsize=13, fontweight="bold", y=1.01
)
plt.tight_layout()
plt.savefig("hinh_3_2_4_stationarity_transformation.png", bbox_inches="tight")
plt.show()
```

### 3.3.1. Liên kết Điểm ngoại lai (Outliers) với Thành phần Nhiễu ($\varepsilon_t$) và Ứng dụng biến đổi Logarit / Symlog

* **Kết nối Điểm ngoại lai (Outliers) và Thành phần Nhiễu ngẫu nhiên (Noise - $\varepsilon_t$):**
  Trong cấu trúc giải phẫu chuỗi thời gian ($Y_t = \text{Signal} + \varepsilon_t$), **Nhiễu ($\varepsilon_t$)** là phần biến động ngẫu nhiên còn lại sau khi đã tách bỏ xu hướng dài hạn và chu kỳ. Các điểm ngoại lai (Outliers) phát hiện qua hàng rào tứ phân vị Tukey (IQR) hoặc điểm chuẩn hóa $|Z\text{-score}| > 2$ trên chuỗi tốc độ tăng trưởng chính là những **cú sốc nhiễu ($\varepsilon_t$) có biên độ cực đại**. Trong bộ dữ liệu tôn giáo toàn cầu, các cú sốc này không phải là lỗi nhập liệu ngẫu nhiên mà phản ánh các biến cố lịch sử, địa chính trị hoặc sự thay đổi phương pháp thống kê nhân khẩu học tại các mốc năm đặc biệt (điển hình là kỳ **1950** ngay sau Thế chiến II và sự phân chia Ấn Độ năm 1947 làm số liệu thống kê Ấn Độ giáo, Hồi giáo, Sikh giáo tăng vọt; hay kỳ **1990–1995** khi Liên Xô tan rã).

* **Vai trò của biến đổi Logarit ($\ln(y_t)$) và thang đo Symlog trong xử lý Outliers:**
  Khi chuỗi thời gian xuất hiện các điểm dị biệt (Outliers) có biên độ quá lớn, phương sai của chuỗi sẽ bị phình to và thay đổi theo thời gian (Heteroscedasticity), làm méo mó các kiểm định thống kê cũng như mô hình dự báo. Việc áp dụng phép biến đổi **Logarit** $y_{\log} = \ln(y_t)$ (hoặc **Log-Return** $\Delta \ln(y_t)$ và thang đo **Symlog** trên Hình 3.3b) đóng vai trò như một bộ giảm chấn toán học: giúp **nén mạnh biên độ của các điểm ngoại lai**, thu hẹp khoảng trải giữa (IQR) và ổn định phương sai của toàn bộ chuỗi dữ liệu trước khi đưa vào mô hình hóa.

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import numpy as np
import pandas as pd
from statsmodels.nonparametric.smoothers_lowess import lowess

# Thiết lập cấu hình hiển thị chuẩn
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. CHUẨN BỊ DỮ LIỆU TỪ df_clean
if "df_clean" in globals():
    df_out_ts = df_clean.copy()
else:
    url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
    df_out_ts = pd.read_csv(url)
    df_out_ts.columns = df_out_ts.columns.str.lower().str.strip()

df_out_ts["year"] = pd.to_numeric(df_out_ts["year"], errors="coerce")
df_out_ts = df_out_ts.sort_values("year").reset_index(drop=True)

relig_cols = {
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo",
    "nonrelig": "Không tôn giáo",
    "sikhgen": "Sikh giáo",
    "judgen": "Do Thái giáo",
    "shntgen": "Thần đạo",
}

for col in relig_cols.keys():
    if col in df_out_ts.columns:
        if df_out_ts[col].dtype == object:
            df_out_ts[col] = df_out_ts[col].astype(str).str.replace(",", "", regex=False).str.strip()
        df_out_ts[col] = pd.to_numeric(df_out_ts[col], errors="coerce").fillna(0)

# 2. TRÍCH XUẤT THÀNH PHẦN NHIỄU (NOISE) VÀ ĐÁNH GIÁ HIỆU QUẢ NÉN PHƯƠNG SAI CỦA LOGARIT
years = df_out_ts["year"].values
summary_outliers = []
noise_store = {}

for col, name in relig_cols.items():
    if col not in df_out_ts.columns:
        continue
    y_raw = df_out_ts[col].values / 1e6  # Đơn vị: Triệu người
    y_log = np.log(np.maximum(y_raw, 1e-6))

    # Tách Signal (Mô hình mượt LOWESS) và Nhiễu (Noise = y_raw - Signal)
    signal_raw = lowess(y_raw, years, frac=0.45, return_sorted=False)
    noise_raw = y_raw - signal_raw
    noise_z = (noise_raw - np.mean(noise_raw)) / (np.std(noise_raw, ddof=1) + 1e-9)

    # Tốc độ tăng trưởng liên kỳ (%) vs Log-Return (%)
    pct_growth = pd.Series(y_raw).pct_change().dropna().values * 100
    log_return = pd.Series(y_log).diff().dropna().values * 100
    growth_years = years[1:]

    # Phát hiện năm ngoại lai theo hàng rào IQR Tukey trên tốc độ tăng trưởng
    q1, q3 = np.percentile(pct_growth, [25, 75])
    iqr_raw = q3 - q1
    out_mask = (pct_growth < (q1 - 1.5 * iqr_raw)) | (pct_growth > (q3 + 1.5 * iqr_raw))
    out_yrs_list = [str(int(yr)) for yr in growth_years[out_mask]]

    # Năm có cú sốc Nhiễu lớn nhất
    max_noise_idx = int(np.argmax(np.abs(noise_z)))
    shock_year = int(years[max_noise_idx])
    shock_z = noise_z[max_noise_idx]

    var_raw_growth = np.var(pct_growth, ddof=1)
    var_log_return = np.var(log_return, ddof=1)

    noise_store[col] = {
        "name": name,
        "years": years,
        "growth_years": growth_years,
        "y_raw": y_raw,
        "signal_raw": signal_raw,
        "noise_raw": noise_raw,
        "noise_z": noise_z,
        "pct_growth": pct_growth,
        "log_return": log_return,
        "out_mask": out_mask,
    }

    summary_outliers.append({
        "Tôn giáo": name,
        "Năm Outliers (Tukey IQR)": ", ".join(out_yrs_list) if out_yrs_list else "Không có",
        "Đỉnh Nhiễu (Noise) lớn nhất": f"Năm {shock_year} (Z = {shock_z:+.2f}s)",
        "Cực đại Tăng trưởng gốc (%)": f"{np.max(np.abs(pct_growth)):.1f}%",
        "Cực đại sau Log-Return (%)": f"{np.max(np.abs(log_return)):.1f}%",
        "Phương sai Tăng trưởng gốc": f"{var_raw_growth:,.1f}",
        "Phương sai sau Log-Return": f"{var_log_return:,.1f}",
    })

df_outlier_noise_table = pd.DataFrame(summary_outliers)

print("==============================================================================================================================")
print(" BẢNG 3.3.1: ĐỐI CHIẾU ĐIỂM NGOẠI LAI (OUTLIERS), ĐỈNH NHIỄU NGẪU NHIÊN VÀ HIỆU QUẢ ỔN ĐỊNH PHƯƠNG SAI BẰNG LOGARIT")
print("==============================================================================================================================")
display(df_outlier_noise_table)

# 3. TRỰC QUAN HÓA: (A) TÁCH SIGNAL & NHIỄU KÈM OUTLIERS | (B) HIỆU QUẢ NÉN OUTLIERS CỦA LOGARIT
fig, axes = plt.subplots(2, 2, figsize=(15, 9.5))

showcase_noise = [("islmgen", "#27ae60", axes[0, 0]), ("nonrelig", "#c0392b", axes[1, 0])]

for col, color, ax in showcase_noise:
    d = noise_store[col]
    yrs = d["years"]
    bar_colors = ["#e74c3c" if abs(z) >= 1.5 else "#95a5a6" for z in d["noise_z"]]
    ax.bar(yrs, d["noise_raw"], width=2.8, color=bar_colors, edgecolor="black", linewidth=0.6, alpha=0.88,
           label="Nhiễu (Noise = Y_t - Signal)")
    ax.axhline(0, color="black", lw=1.2)

    sd_noise = np.std(d["noise_raw"], ddof=1)
    ax.axhline(1.5 * sd_noise, color="red", linestyle="--", lw=1.1, label="Ngưỡng dị biệt (+- 1.5s)")
    ax.axhline(-1.5 * sd_noise, color="red", linestyle="--", lw=1.1)

    for y_val, n_val, z_val in zip(yrs, d["noise_raw"], d["noise_z"]):
        if abs(z_val) >= 1.5:
            ax.annotate(
                f"Năm {int(y_val)}\n(Nhiễu = {n_val:+.1f} tr)",
                xy=(y_val, n_val),
                xytext=(0, 10 if n_val >= 0 else -24),
                textcoords="offset points",
                ha="center", fontsize=8.5, fontweight="bold", color="darkred",
            )

    ax.set_title(f"a) Giải phẫu Nhiễu & Cú sốc Ngoại lai: {d['name'].upper()}",
                 fontsize=10.5, fontweight="bold")
    ax.set_ylabel("Nhiễu (Triệu người)", fontweight="bold")
    ax.set_xlabel("Năm khảo sát", fontweight="bold")
    ax.xaxis.set_major_locator(mt.MultipleLocator(10))
    ax.legend(loc="upper left", frameon=True, fontsize=8.5)
    ax.grid(True, linestyle=":", alpha=0.6)

# CỘT PHẢI HÀNG 1: So sánh Tốc độ tăng trưởng Gốc (%) vs Log-Return tại kỳ đột biến 1950
shock_cols = ["hindgen", "sikhgen", "islmgen", "nonrelig", "chrstgen", "budgen"]
labels_comp = [noise_store[c]["name"] for c in shock_cols if c in noise_store]
max_pct_vals = [np.max(np.abs(noise_store[c]["pct_growth"])) for c in shock_cols if c in noise_store]
max_log_vals = [np.max(np.abs(noise_store[c]["log_return"])) for c in shock_cols if c in noise_store]

x_pos = np.arange(len(labels_comp))
width = 0.36
ax_b1 = axes[0, 1]
rects1 = ax_b1.bar(x_pos - width / 2, max_pct_vals, width, label="Biên độ gốc (%)", color="#e67e22", edgecolor="black")
rects2 = ax_b1.bar(x_pos + width / 2, max_log_vals, width, label="Sau biến đổi Log (%)", color="#1f4e79", edgecolor="black")

ax_b1.set_title("b) Hiệu quả nén biên độ Outliers của hàm Logarit (Kỳ đột biến 1950)", fontsize=10.5, fontweight="bold")
ax_b1.set_xticks(x_pos)
ax_b1.set_xticklabels(labels_comp, rotation=20, ha="right", fontweight="bold", fontsize=9)
ax_b1.set_ylabel("Biên độ cực đại (%)", fontweight="bold")
ax_b1.legend(loc="upper right", frameon=True, fontsize=8.5)
ax_b1.grid(axis="y", linestyle=":", alpha=0.6)

# CỘT PHẢI HÀNG 2: Minh họa sự ổn định phương sai trên chuỗi Ấn Độ giáo
ax_b2 = axes[1, 1]
if "hindgen" in noise_store:
    d_hind = noise_store["hindgen"]
    ax_b2.plot(d_hind["growth_years"], d_hind["pct_growth"], marker="o", color="#c0392b", lw=2,
               label=f"Tăng trưởng gốc (Outlier 1950 = +{d_hind['pct_growth'][0]:.0f}%)")
    ax_b2.plot(d_hind["growth_years"], d_hind["log_return"], marker="s", color="#117a65", lw=2.2, linestyle="--",
               label=f"Log-Return (Biên độ giảm còn: +{d_hind['log_return'][0]:.0f}%)")
ax_b2.set_title("c) Sự ổn định phương sai và nén nhiễu trên chuỗi Ấn Độ giáo", fontsize=10.5, fontweight="bold")
ax_b2.set_ylabel("Biến động liên kỳ (%)", fontweight="bold")
ax_b2.set_xlabel("Năm khảo sát", fontweight="bold")
ax_b2.xaxis.set_major_locator(mt.MultipleLocator(10))
ax_b2.legend(loc="upper right", frameon=True, fontsize=8.5)
ax_b2.grid(True, linestyle=":", alpha=0.6)

plt.suptitle("HÌNH 3.3.1: PHÂN TÍCH TÁC ĐỘNG CỦA OUTLIERS VÀ HIỆU QUẢ CỦA BIẾN ĐỔI LOGARIT CHUỖI THỜI GIAN",
             fontsize=12, fontweight="bold", y=1.02)
plt.tight_layout()
plt.savefig("hinh_3_3_1_outliers_noise_analysis.png", bbox_inches="tight")
plt.show()
```

### 3.4 Phân tích mối tương quan (Correlation Analysis)


*   **Nội dung**: Phân tích mối tương quan giữa các biến định lượng, chẳng hạn như tổng dân số toàn cầu (worldpop), quy mô số lượng tín đồ và tỷ lệ phần trăm (pct) của các nhóm tôn giáo. Sử dụng hệ số tương quan Pearson hoặc Spearman tùy thuộc vào đặc điểm phân phối của dữ liệu.

*   **Mục đích**: Xác định mức độ và chiều hướng liên hệ giữa các biến số, hỗ trợ tìm ra những xu hướng dịch chuyển nhân khẩu học đáng chú ý trong bộ dữ liệu chuỗi thời gian.



```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import pandas as pd
import seaborn as sns

# Thiết lập hiển thị chuẩn cho báo cáo học thuật
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. ĐỌC VÀ CHUẨN BỊ DỮ LIỆU
if "df_clean" in globals():
  df_corr = df_clean.copy()
else:
  url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
  df_corr = pd.read_csv(url)
  df_corr.columns = df_corr.columns.str.lower().str.strip()

# Các biến cần phân tích tương quan
major_abs_cols = [
    "worldpop",
    "chrstgen",
    "islmgen",
    "hindgen",
    "budgen",
    "nonrelig",
]
religion_rename_map = {
    "worldpop": "Dân số thế giới",
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo",
    "nonrelig": "Không tôn giáo",
}

# Làm sạch và ép kiểu dữ liệu
for col in major_abs_cols:
  if col in df_corr.columns:
    if df_corr[col].dtype == object:
      df_corr[col] = (
          df_corr[col]
          .astype(str)
          .str.replace(",", "", regex=False)
          .str.strip()
      )
    df_corr[col] = pd.to_numeric(df_corr[col], errors="coerce").fillna(0)

# Lấy dữ liệu và đổi tên sang tiếng Việt
df_sub = df_corr[major_abs_cols].rename(columns=religion_rename_map)

# 2. KHỞI TẠO KHUNG BIỂU ĐỒ (1 HÀNG 2 CỘT)
fig, axes = plt.subplots(1, 2, figsize=(15, 6.5))

# =============================================================
# BIỂU ĐỒ a: HEATMAP MA TRẬN TƯƠNG QUAN PEARSON
# =============================================================
# Tính ma trận tương quan Pearson
corr_matrix = df_sub.corr(method="pearson")

sns.heatmap(
    corr_matrix,
    annot=True,  # Hiển thị số liệu
    fmt=".3f",  # Làm tròn 3 chữ số thập phân
    cmap="RdYlBu_r",  # Thang màu từ đỏ (tương quan âm) sang xanh (tương quan dương)
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    linecolor="white",
    cbar_kws={"label": "Hệ số tương quan Pearson"},
    ax=axes[0],
)

axes[0].set_title(
    "a) Ma trận tương quan Pearson giữa các tôn giáo & Dân số",
    fontsize=11.5,
    fontweight="bold",
    pad=12,
)
axes[0].set_xticklabels(
    axes[0].get_xticklabels(), rotation=30, ha="right", fontweight="bold"
)
axes[0].set_yticklabels(
    axes[0].get_yticklabels(), rotation=0, fontweight="bold"
)

# =============================================================
# BIỂU ĐỒ b: SCATTER PLOT KÈM ĐƯỜNG HỒI QUY (TUYẾN TÍNH)
# =============================================================
# Biểu diễn Dân số thế giới với 2 tôn giáo lớn nhất
sns.regplot(
    data=df_sub,
    x="Dân số thế giới",
    y="Cơ Đốc giáo",
    label="Cơ Đốc giáo",
    color="#1f4e79",
    marker="o",
    scatter_kws={"s": 50, "alpha": 0.8},
    line_kws={"linewidth": 2},
    ax=axes[1],
)

sns.regplot(
    data=df_sub,
    x="Dân số thế giới",
    y="Hồi giáo",
    label="Hồi giáo",
    color="#27ae60",
    marker="s",
    scatter_kws={"s": 50, "alpha": 0.8},
    line_kws={"linewidth": 2, "linestyle": "--"},
    ax=axes[1],
)

# Định dạng các trục hiển thị theo tỷ người (Billion)
formatter = mt.FuncFormatter(lambda v, _: f"{v / 1e9:.1f} tỷ")
axes[1].xaxis.set_major_formatter(formatter)
axes[1].yaxis.set_major_formatter(formatter)

axes[1].set_xlabel("Tổng dân số thế giới (worldpop)", fontweight="bold")
axes[1].set_ylabel("Số lượng tín đồ", fontweight="bold")
axes[1].set_title(
    "b) Tương quan gia tăng tín đồ theo quy mô dân số toàn cầu",
    fontsize=11.5,
    fontweight="bold",
    pad=12,
)
axes[1].grid(True, linestyle="--", alpha=0.5)
axes[1].legend(title="Tôn giáo", frameon=True, facecolor="white")

plt.suptitle(
    "HÌNH 3.4: PHÂN TÍCH TƯƠNG QUAN VÀ XU HƯỚNG BIẾN ĐỘNG DÂN SỐ TÔN GIÁO (1945–2010)",
    fontsize=13,
    fontweight="bold",
    y=1.02,
)
plt.tight_layout()
plt.savefig("hinh_3_4_correlation_analysis.png", bbox_inches="tight")
plt.show()
```

#### 3.4.1: Sử dụng Ma trận phân tán (Pairplot) để trực quan hóa mối liên hệ đa biến giữa các nhóm tôn giáo

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Thiết lập cấu hình
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

print("--- PHÂN TÍCH TƯƠNG QUAN CHÉO QUY MÔ 4 TÔN GIÁO LỚN ---")

# 1. LẤY VÀ CHUẨN BỊ DỮ LIỆU
relig_cols = ["chrstgen", "islmgen", "hindgen", "budgen"]
relig_names = ["Cơ Đốc giáo", "Hồi giáo", "Ấn Độ giáo", "Phật giáo"]

# Sử dụng df_clean từ bộ nhớ Colab
if "df_clean" in globals():
    df_cross = df_clean[["year"] + relig_cols].copy()
    # Làm sạch và đổi đơn vị sang Triệu người cho dễ nhìn
    for col in relig_cols:
        df_cross[col] = pd.to_numeric(df_cross[col].astype(str).str.replace(",", "").str.strip(), errors="coerce") / 1e6
    df_cross.columns = ["Năm"] + relig_names
else:
    print("Lỗi: Không tìm thấy df_clean. Vui lòng chạy lại phần Đọc dữ liệu.")

# 2. TÍNH MA TRẬN TƯƠNG QUAN THỨ BẬC SPEARMAN
corr_matrix = df_cross[relig_names].corr(method="spearman")

# 3. TRỰC QUAN HÓA BẰNG PAIRPLOT (MA TRẬN PHÂN TÁN)
# Vẽ lưới đồ thị: Đường chéo là biểu đồ phân phối (KDE), các ô còn lại là Scatter plot kèm đường hồi quy
g = sns.pairplot(
    df_cross[relig_names],
    kind="reg",
    diag_kind="kde",
    corner=True, # Chỉ vẽ góc dưới bên trái để tránh trùng lặp
    plot_kws={'line_kws': {'color': 'red', 'linewidth': 2}, 'scatter_kws': {'alpha': 0.6, 'color': '#1f4e79', 's': 40}},
    diag_kws={'color': '#27ae60', 'fill': True, 'alpha': 0.5}
)

g.fig.suptitle(
    "HÌNH 3.4.2: MA TRẬN PHÂN TÁN VÀ TƯƠNG QUAN CHÉO\nGIỮA QUY MÔ 4 TÔN GIÁO LỚN (ĐV: Triệu người)",
    y=1.05, fontweight="bold", fontsize=12
)

plt.savefig("hinh_3_4_2_cross_correlation.png", bbox_inches="tight")
plt.show()

# 4. IN RA KẾT LUẬN DỰA TRÊN KẾT QUẢ MA TRẬN
print("-" * 80)
print("KẾT LUẬN PHÂN TÍCH TƯƠNG QUAN CHÉO:")

# Trích xuất hệ số tương quan trung bình
avg_corr = corr_matrix.unstack().sort_values().drop_duplicates()
avg_corr = avg_corr[avg_corr < 1.0].mean()

if avg_corr > 0.8:
    print(f"1. Chiều hướng: Biểu đồ Scatter Plot và đường hồi quy (màu đỏ) đều hướng lên, cùng với hệ số tương quan trung bình giữa các cặp đạt mức rất cao (ρ ≈ {avg_corr:.2f}).")
    print("2. Đánh giá tính đồng pha: 4 tôn giáo lớn không triệt tiêu lẫn nhau (không có tương quan âm). Thay vào đó, chúng tăng trưởng hoàn toàn đồng pha.")
    print("3. Yếu tố chi phối: Mối tương quan thuận cực mạnh này chứng minh rằng động lực tăng trưởng tuyệt đối của cả 4 tôn giáo đều bị chi phối chung bởi một biến số vĩ mô độc lập: Sự bùng nổ dân số toàn cầu (World Population) từ năm 1945 đến 2010.")
else:
    print("Có sự phân hóa trong tốc độ phát triển giữa các tôn giáo, một số tăng trưởng trong khi nhóm khác chững lại hoặc suy giảm.")
print("-" * 80)
```

### 3.5 Trực quan hóa dữ liệu (Data Visualization)


*   **Nội dung**: Ứng dụng các loại biểu đồ chuyên dụng như Bar Chart, Line Chart, Boxplot và Scatter Plot để minh họa sự phân bố, cơ cấu và sự thay đổi của dữ liệu theo thời gian (các mốc định kỳ từ 1945 đến 2010).

*   **Mục đích**: Biến đổi các con số khô khan thành hình ảnh trực quan, giúp dễ dàng quan sát xu hướng tăng trưởng, sự dịch chuyển tỷ trọng giữa các tôn giáo và phát hiện nhanh chóng các biến động bất thường.



```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mt
import pandas as pd
import seaborn as sns

# Thiết lập cấu hình hiển thị chuẩn
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

# 1. KẾ THỪA DỮ LIỆU TỪ df_clean CỦA CHƯƠNG 2
if "df_clean" in globals():
  df_viz = df_clean.copy()
else:
  url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
  df_viz = pd.read_csv(url)
  df_viz.columns = df_viz.columns.str.lower().str.strip()

# Ánh xạ tên tôn giáo
relig_map = {
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo",
    "nonrelig": "Không tôn giáo",
}

valid_cols = [c for c in relig_map.keys() if c in df_viz.columns]

# Ép kiểu dữ liệu an toàn
for c in valid_cols:
  if df_viz[c].dtype == object:
    df_viz[c] = df_viz[c].astype(str).str.replace(",", "", regex=False).str.strip()
  df_viz[c] = pd.to_numeric(df_viz[c], errors="coerce").fillna(0)

# Tính tỷ lệ % trên tổng số
total_pop = df_viz[valid_cols].sum(axis=1)
df_pct = pd.DataFrame()
for c in valid_cols:
  df_pct[relig_map[c]] = (df_viz[c] / total_pop) * 100

# Chuyển dữ liệu sang dạng Long-form
df_melted = df_pct.melt(var_name="Tôn giáo", value_name="Tỷ lệ (%)")

# 2. VẼ BIỂU ĐỒ VIOLIN PLOT KẾT HỢP BOXPLOT
fig, ax = plt.subplots(figsize=(11, 6.5))

# Vẽ Violin Plot để thấy hình dáng phân phối và mật độ dữ liệu
sns.violinplot(
    data=df_melted,
    x="Tôn giáo",
    y="Tỷ lệ (%)",
    hue="Tôn giáo",
    inner=None,
    palette="Pastel1",
    legend=False,
    ax=ax,
    linewidth=1.2,
)

# Lồng Boxplot bên trong để hiển thị chính xác các mốc tứ phân vị (Q1, Trung vị, Q3)
sns.boxplot(
    data=df_melted,
    x="Tôn giáo",
    y="Tỷ lệ (%)",
    width=0.15,
    color="white",
    boxprops=dict(zorder=2),
    ax=ax,
)

# Định dạng biểu đồ
ax.set_title(
    "HÌNH 3.5: PHÂN PHỐI TỶ TRỌNG CÁC TÔN GIÁO LỚN QUA VIOLIN & BOXPLOT (1945–2010)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax.set_xlabel("Nhóm tôn giáo", fontweight="bold", fontsize=11)
ax.set_ylabel("Tỷ trọng trong tổng dân số thế giới (%)", fontweight="bold", fontsize=11)
ax.yaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:.0f}%"))
ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("hinh_3_5_violin_boxplot_fixed.png", bbox_inches="tight")
plt.show()
```

#### 3.5.1: Phân tích tính độc lập (Chi-Squared Test): Đánh giá mối liên hệ giữa nhóm tôn giáo và tốc độ tăng trưởng

```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import chi2_contingency
from statsmodels.graphics.mosaicplot import mosaic
from matplotlib.patches import Patch

# Thiết lập cấu hình hiển thị chuẩn
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

print("=====================================================================")
print("  MỤC 3.5: PHÂN TÍCH TÍNH ĐỘC LẬP (CHI-SQUARED & MOSAIC PLOT)")
print("=====================================================================\n")

# 1. CHUẨN BỊ DỮ LIỆU
if "df_clean" in globals():
    df_chi = df_clean.copy()
else:
    url = "https://raw.githubusercontent.com/28-TranTrungKien-24174600028/Do_An_1_Nhom_2_3_DHKL18A1HN/refs/heads/main/WRP%20global%20data.csv"
    df_chi = pd.read_csv(url)
    df_chi.columns = df_chi.columns.str.lower().str.strip()

names_map = {
    "chrstgen": "Cơ Đốc giáo",
    "islmgen": "Hồi giáo",
    "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo"
}

# Ép kiểu dữ liệu
for col in names_map.keys():
    if col in df_chi.columns:
        if df_chi[col].dtype == object:
            df_chi[col] = df_chi[col].astype(str).str.replace(',', '', regex=False).str.strip()
        df_chi[col] = pd.to_numeric(df_chi[col], errors='coerce').fillna(0)

# Tính tốc độ tăng trưởng liên kỳ 5 năm (%)
df_growth = df_chi.set_index("year")[list(names_map.keys())].pct_change().dropna() * 100
df_growth.columns = [names_map[c] for c in df_growth.columns]

# Chuyển đổi dữ liệu sang dạng Long-form
df_long = df_growth.reset_index().melt(id_vars="year", var_name="Tôn giáo", value_name="Tốc độ (%)")

# Tạo biến phân loại thứ 2: Phân nhóm tăng trưởng
df_long["Nhóm tăng trưởng"] = df_long["Tốc độ (%)"].apply(lambda x: "Tăng trưởng cao (>10%)" if x >= 10 else "Thấp / Giảm (<10%)")

# 2. KIỂM ĐỊNH CHI-SQUARED
contingency_table = pd.crosstab(df_long["Tôn giáo"], df_long["Nhóm tăng trưởng"])
chi2, p_val, dof, expected = chi2_contingency(contingency_table)

# 3. TRỰC QUAN HÓA BẰNG MOSAIC PLOT
fig, ax = plt.subplots(figsize=(10, 6))

# Định nghĩa màu sắc: Xanh cho tăng trưởng cao, Đỏ cho thấp/giảm
props = lambda key: {'color': '#27ae60' if 'Tăng trưởng cao' in key[1] else '#c0392b'}
labelizer = lambda k: "" # Ẩn text mặc định

mosaic(
    df_long,
    ['Tôn giáo', 'Nhóm tăng trưởng'],
    ax=ax,
    properties=props,
    labelizer=labelizer,
    gap=0.02
)

# ĐÃ SỬA LỖI Ở ĐÂY: Thêm chữ 'r' trước f-string để Python hiểu đúng ký tự LaTeX \chi
ax.set_title(
    f"HÌNH 3.5: BIỂU ĐỒ LÁT GẠCH (MOSAIC PLOT) - TÔN GIÁO vs MỨC ĐỘ TĂNG TRƯỞNG\n"
    rf"(Kiểm định Chi-Squared: $\chi^2$ = {chi2:.2f}, p-value = {p_val:.4f})",
    fontsize=12, fontweight="bold", pad=20
)

# Thêm chú thích (Legend)
legend_elements = [
    Patch(facecolor='#27ae60', edgecolor='white', label='Tăng trưởng cao (>10% mỗi 5 năm)'),
    Patch(facecolor='#c0392b', edgecolor='white', label='Thấp / Giảm (<10% mỗi 5 năm)')
]
ax.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(1.02, 1), frameon=False, title="Nhóm tăng trưởng", title_fontproperties={'weight': 'bold'})

plt.tight_layout()
plt.savefig("hinh_3_5_chi2_mosaic_plot.png", bbox_inches="tight")
plt.show()

# 4. IN RA BẢNG VÀ KẾT LUẬN THỐNG KÊ
print("--- 1. BẢNG TẦN SỐ QUAN SÁT (CONTINGENCY TABLE) ---")
print(contingency_table)
print("\n--- 2. KẾT QUẢ KIỂM ĐỊNH CHI-SQUARED ---")
print(f"- Chi-Squared (X2): {chi2:.2f}")
print(f"- P-value: {p_val:.4f}")
print(f"- Bậc tự do (dof): {dof}")

print("\n--- 3. KẾT LUẬN VÀ GIẢI THÍCH ---")
if p_val < 0.05:
    print("-> BÁC BỎ giả thuyết H0 (p-value < 0.05).")
    print("-> Giải thích: Có mối liên hệ phụ thuộc (có ý nghĩa thống kê) giữa đặc tính của từng tôn giáo và khả năng lọt vào nhóm tăng trưởng cao.")
    print("   Nhìn vào biểu đồ Mosaic, Hồi giáo phá vỡ quy luật độc lập khi có khối lượng cột màu xanh (Tăng trưởng cao) áp đảo diện tích, trong khi Cơ Đốc giáo lại có màu đỏ chiếm ưu thế.")
else:
    print("-> CHẤP NHẬN giả thuyết H0 (p-value >= 0.05).")
    print("-> Giải thích: Hai biến độc lập với nhau. Nhóm tôn giáo không ảnh hưởng đến việc phân loại tốc độ tăng trưởng. Các tôn giáo đều có chu kỳ tăng/giảm ngẫu nhiên và tương đồng nhau trước các tác động của bối cảnh toàn cầu.")
print("=====================================================================")
```

### 4.1 : Trực quan hóa quy mô & cấu trúc tổng thể


*   **Mục đích**: Đánh giá tổng quan quy mô tín đồ tuyệt đối và tỷ lệ phần trăm của các nhóm tôn giáo chính (Cơ Đốc giáo, Hồi giáo, Phật giáo, Ấn Độ giáo...) so với tổng dân số thế giới (worldpop) tại các mốc thời gian cụ thể (ví dụ: năm 1945 và năm 2010).

*   **Cách thực hiện**: Sử dụng Bar Chart (Biểu đồ cột) để so sánh quy mô tín đồ tuyệt đối giữa các tôn giáo; sử dụng Pie Chart hoặc Donut Chart để thể hiện tỷ lệ phần trăm (pct) của từng nhóm tôn giáo trong tổng dân số toàn cầu.

*   **Mục đích cụ thể**: Giúp nhận diện nhanh chóng tôn giáo nào đang chiếm đa số và cấu trúc tôn giáo toàn cầu tại các thời điểm lịch sử.




```python
# ## 3.1 Trực quan hóa quy mô & cấu trúc tổng thể
# Sử dụng trực tiếp df_clean đã được xử lý từ Chương 2 để vẽ biểu đồ

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mt

# Thiết lập cấu hình vẽ biểu đồ
plt.rcParams.update({"font.family": "DejaVu Sans", "figure.dpi": 120,
                     "axes.spines.top": False, "axes.spines.right": False})

# Tận dụng df_clean đã xử lý sẵn từ Chương 2 (Không dùng phương án tải lại dự phòng)
df = df_clean.copy()
df = df.set_index("year")

# Ánh xạ nhóm tôn giáo sang tiếng Việt
groups = {
    "chrstgen": "Cơ Đốc giáo", "islmgen": "Hồi giáo", "hindgen": "Ấn Độ giáo",
    "budgen": "Phật giáo", "judgen": "Do Thái giáo", "sikhgen": "Sikh giáo",
    "shntgen": "Thần đạo", "bahgen": "Baha'i", "taogen": "Đạo giáo",
    "jaingen": "Kỳ Na giáo", "confgen": "Nho giáo", "syncgen": "Tôn giáo hỗn hợp",
    "anmgen": "Vật linh giáo", "zorogen": "Bái hỏa giáo", "othrgen": "Tôn giáo khác",
    "nonrelig": "Không tôn giáo",
}
YEARS = [1945, 2010]
available_groups = {k: v for k, v in groups.items() if k in df.columns}

# Đảm bảo các cột giá trị định lượng là kiểu số (Làm sạch dấu phẩy trước khi chuyển đổi)
for col in list(available_groups) + ["sumrelig", "worldpop"]:
    if col in df.columns:
        if df[col].dtype == object:
            df[col] = df[col].astype(str).str.replace(',', '', regex=False).str.strip()
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0)

abs_df = df.loc[YEARS, list(available_groups)].rename(columns=available_groups).T
abs_df = abs_df.sort_values(2010, ascending=False)
total = df.loc[YEARS, "sumrelig"]
share_df = abs_df / total * 100

print("Tổng dân số trong dữ liệu (sumrelig):\n", total)
print("worldpop:\n", df.loc[YEARS, "worldpop"])
print("Độ phủ sumrelig/worldpop (%):\n", (total / df.loc[YEARS, "worldpop"] * 100).round(1))

summary = pd.concat({"Số tín đồ": abs_df, "Tỷ lệ %": share_df.round(2)}, axis=1)
display(summary)

# --- Biểu đồ 1: Bar chart số tín đồ tuyệt đối (triệu người) ---
col_colors = {1945: "#9bb7d4", 2010: "#1f4e79"}
plot_df = (abs_df / 1e6)
fig, ax = plt.subplots(figsize=(11, 7))
y = range(len(plot_df))
h = 0.4
for i, yr in enumerate(YEARS):
    bars = ax.barh([p + (i - 0.5) * h for p in y], plot_df[yr], height=h,
                   label=str(yr), color=col_colors[yr])
    for b, v in zip(bars, plot_df[yr]):
        ax.text(v + 15, b.get_y() + b.get_height() / 2, f"{v:,.0f}", va="center", fontsize=7.5)
ax.set_yticks(list(y))
ax.set_yticklabels(plot_df.index)
ax.invert_yaxis()
ax.xaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:,.0f}"))
ax.set_xlabel("Số tín đồ (triệu người)")
ax.set_title("Quy mô tín đồ tuyệt đối theo nhóm tôn giáo: 1945 so với 2010", fontweight="bold")
ax.legend(title="Năm")
plt.tight_layout()
plt.savefig("hinh_3_1a_bar_tin_do_tuyet_doi.png", bbox_inches="tight")
plt.show()

# --- Biểu đồ 2: Donut chart tỷ lệ % (gộp nhóm < 1.5% thành "Khác") ---
def to_donut(series, thr=1.5):
    big = series[series >= thr]
    small = series[series < thr].sum()
    return pd.concat([big, pd.Series({"Các nhóm nhỏ khác": small})])

order = share_df[2010].index.tolist()
palette = dict(zip(order, plt.cm.tab20.colors[:len(order)]))
palette["Các nhóm nhỏ khác"] = "#bdbdbd"

fig, axes = plt.subplots(1, 2, figsize=(14, 7))
for ax, yr in zip(axes, YEARS):
    s = to_donut(share_df[yr])
    wedges, _ = ax.pie(s, colors=[palette.get(k, "#bdbdbd") for k in s.index], startangle=90,
                       counterclock=False, wedgeprops=dict(width=0.38, edgecolor="white"))
    ax.set_title(f"Năm {yr}", fontsize=13, fontweight="bold")
    ax.text(0, 0, f"{total[yr]/1e9:.2f} tỷ\nngười", ha="center", va="center", fontsize=12)
    ax.legend(wedges, [f"{k}: {v:.1f}%" for k, v in s.items()],
              loc="upper center", bbox_to_anchor=(0.5, -0.02), ncol=2, fontsize=9, frameon=False)
fig.suptitle("Cơ cấu tôn giáo toàn cầu (% tổng dân số)", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.savefig("hinh_3_1b_donut_ty_le.png", bbox_inches="tight")
plt.show()

# --- Biểu đồ 3: Thay đổi tỷ lệ (điểm phần trăm) ---
def_colors = ["#c0392b" if v < 0 else "#27ae60" for v in (share_df[2010] - share_df[1945]).sort_values()]
delta = (share_df[2010] - share_df[1945]).sort_values()
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(delta.index, delta, color=def_colors)
for i, v in enumerate(delta):
    ax.text(v + (0.3 if v >= 0 else -0.3), i, f"{v:+.1f}", va="center",
            ha="left" if v >= 0 else "right", fontsize=8)
ax.axvline(0, color="black", lw=0.8)
ax.set_xlabel("Thay đổi tỷ lệ (điểm phần trăm)")
ax.set_title("Thay đổi cơ cấu tôn giáo 1945 → 2010", fontweight="bold")
plt.tight_layout()
plt.savefig("hinh_3_1c_thay_doi_ty_le.png", bbox_inches="tight")
plt.show()
```

### 4.2 Trực quan hóa xu hướng tăng trưởng (Line Chart)


*   **Mục đích**: Theo dõi sự biến động và tốc độ tăng trưởng về mặt quy mô tín đồ của các tôn giáo lớn trải dài qua các thập kỷ (từ 1945 đến 2010).

*   **Biểu đồ sử dụng**: Line Chart (Biểu đồ đường) với nhiều đường trên cùng một trục thời gian (year), mỗi đường đại diện cho tổng số tín đồ của một tôn giáo lớn (như chrstgen, islmgen, budgen, hindgen).

*   **Mục đích cụ thể**: Xác định tốc độ tăng trưởng, nhận diện tôn giáo nào đang phát triển bùng nổ, tôn giáo nào đang chững lại hoặc suy giảm quy mô qua các thời kỳ.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mt

# Thiết lập cấu hình hiển thị biểu đồ
plt.rcParams.update({"font.family": "DejaVu Sans", "figure.dpi": 120,
                      "axes.spines.top": False, "axes.spines.right": False})

# Sao chép và chuẩn hóa dữ liệu dòng thời gian trực tiếp từ df_clean của Chương 2
df_line = df_clean.copy()
df_line['year'] = pd.to_numeric(df_line['year'], errors='coerce')

# Danh sách các tôn giáo lớn cần vẽ xu hướng (chuyển đổi đơn vị sang triệu người cho dễ quan sát)
major_religions = {
    'chrstgen': 'Cơ Đốc giáo',
    'islmgen': 'Hồi giáo',
    'hindgen': 'Ấn Độ giáo',
    'budgen': 'Phật giáo',
    'nonrelig': 'Không tôn giáo',
    'shntgen': 'Thần đạo'
}

# Đảm bảo các cột số liệu ở định dạng numeric
for col in major_religions.keys():
    if col in df_line.columns:
        if df_line[col].dtype == object:
            df_line[col] = df_line[col].astype(str).str.replace(',', '', regex=False).str.strip()
        df_line[col] = pd.to_numeric(df_line[col], errors='coerce').fillna(0) / 1e6 # Đổi sang đơn vị Triệu người

# Vẽ biểu đồ Line Chart
fig, ax = plt.subplots(figsize=(11, 6.5))

# Màu sắc riêng biệt cho từng đường tôn giáo
colors = {
    'chrstgen': '#1f4e79',  # Xanh đậm
    'islmgen': '#27ae60',   # Xanh lá
    'hindgen': '#e67e22',   # Cam
    'budgen': '#8e44ad',    # Tím
    'nonrelig': '#e74c3c',  # Đỏ
    'shntgen': '#34495e'    # Xám đậm
}

for col, name in major_religions.items():
    if col in df_line.columns:
        ax.plot(df_line['year'], df_line[col], marker='o', linewidth=2.5,
                markersize=5.5, label=name, color=colors.get(col, '#7f8c8d'))

# Định dạng các trục và tiêu đề
ax.set_title("Xu hướng tăng trưởng quy mô tín đồ các tôn giáo lớn (1945 - 2010)", fontsize=13, fontweight="bold", pad=15)
ax.set_xlabel("Năm", fontsize=11, fontweight="bold")
ax.set_ylabel("Số lượng tín đồ (Triệu người)", fontsize=11, fontweight="bold")

ax.xaxis.set_major_locator(mt.MultipleLocator(5)) # Hiển thị mốc năm chi tiết
ax.yaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:,.0f} tr"))

ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(title="Nhóm tôn giáo", frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()
plt.savefig("hinh_3_2_line_xu_huong_tang_truong.png", bbox_inches="tight")
plt.show()
```

### 4.3 Trực quan hóa sự dịch chuyển tỷ trọng (Stacked Bar Chart)


*   **Mục đích**: Phân tích sự thay đổi cơ cấu và tỷ trọng của các tôn giáo lớn trong tổng dân số thế giới qua từng mốc định kỳ.

*   **Biểu đồ sử dụng**: Stacked Bar Chart (Biểu đồ cột chồng 100%) với trục hoành là các năm (1945, 1950, ..., 2010) và các cột được phân chia màu sắc tương ứng với tỷ lệ phần trăm (pct) của các nhóm tôn giáo.

*   **Mục đích cụ thể**: Quan sát rõ nét sự dịch chuyển "thị phần" tín ngưỡng và cơ cấu nhân khẩu học toàn cầu qua các thời kỳ lịch sử.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mt

# Thiết lập cấu hình hiển thị biểu đồ
plt.rcParams.update({"font.family": "DejaVu Sans", "figure.dpi": 120,
                      "axes.spines.top": False, "axes.spines.right": False})

# Chuẩn bị dữ liệu tỷ trọng cho biểu đồ cột chồng 100% trực tiếp từ df_clean
df_stack = df_clean.copy()
df_stack['year'] = pd.to_numeric(df_stack['year'], errors='coerce')

# Danh sách các tôn giáo lớn
gen_relig_cols = ['chrstgen', 'islmgen', 'hindgen', 'budgen', 'nonrelig', 'shntgen', 'othrgen']
religion_names = {
    'chrstgen': 'Cơ Đốc giáo',
    'islmgen': 'Hồi giáo',
    'hindgen': 'Ấn Độ giáo',
    'budgen': 'Phật giáo',
    'nonrelig': 'Không tôn giáo',
    'shntgen': 'Thần đạo',
    'othrgen': 'Tôn giáo khác'
}

# Lọc các cột thực tế tồn tại trong dataframe
valid_cols = [c for c in gen_relig_cols if c in df_stack.columns]

# Đảm bảo các cột giá trị là kiểu số
for col in valid_cols:
    if df_stack[col].dtype == object:
        df_stack[col] = df_stack[col].astype(str).str.replace(',', '', regex=False).str.strip()
    df_stack[col] = pd.to_numeric(df_stack[col], errors='coerce').fillna(0)

# Tính tổng cộng dân số theo từng năm từ các nhóm trên để quy đổi ra tỷ lệ phần trăm (100%)
df_stack['total_pop'] = df_stack[valid_cols].sum(axis=1)

# Tạo bảng tỷ lệ phần trăm (%)
df_pct = pd.DataFrame({'year': df_stack['year']})
for col in valid_cols:
    df_pct[col] = (df_stack[col] / df_stack['total_pop']) * 100

df_pct = df_pct.set_index('year')

# Vẽ biểu đồ Stacked Bar Chart (100%)
fig, ax = plt.subplots(figsize=(12, 7))

# Định nghĩa bảng màu trực quan cho các nhóm
colors_list = ['#1f4e79', '#27ae60', '#e67e22', '#8e44ad', '#e74c3c', '#34495e', '#7f8c8d']
color_dict = {col: colors_list[i % len(colors_list)] for i, col in enumerate(valid_cols)}

# Vẽ lần lượt từng lớp cột chồng lên nhau
bottom = np.zeros(len(df_pct))
for col in valid_cols:
    values = df_pct[col].values
    label_name = religion_names.get(col, col)
    ax.bar(df_pct.index, values, bottom=bottom, label=label_name,
           color=color_dict.get(col, '#95a5a6'), width=3.5, edgecolor='white', linewidth=0.5)
    bottom += values

# Định dạng biểu đồ
ax.set_title("Sự dịch chuyển cơ cấu và tỷ trọng các tôn giáo toàn cầu (1945 - 2010)", fontsize=13, fontweight="bold", pad=15)
ax.set_xlabel("Năm", fontsize=11, fontweight="bold")
ax.set_ylabel("Tỷ trọng (%)", fontsize=11, fontweight="bold")

ax.set_ylim(0, 100)
ax.yaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:.0f}%"))
ax.xaxis.set_major_locator(mt.MultipleLocator(5))

ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.legend(title="Nhóm tôn giáo", bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()
plt.savefig("hinh_3_3_stacked_bar_ty_trong.png", bbox_inches="tight")
plt.show()
```

### 4.4 Khai phá cấu trúc nội bộ của các nhóm Tôn giáo lớn


*   **Mục đích**: Phân tích sâu hơn vào cấu trúc phân nhánh bên trong của một tôn giáo cụ thể (ví dụ: các nhánh của Cơ Đốc giáo hoặc Hồi giáo).

*   **Biểu đồ sử dụng**: Biểu đồ cột chồng (Stacked Bar Chart) thể hiện tỷ trọng đóng góp của các nhánh nhỏ (như Tin Lành chrstprot, Công giáo chrstcat, Chính thống giáo chrstorth trong tổng số chrstgen).

*   **Mục đích cụ thể**: Hiểu rõ nhánh nào đang đóng vai trò chủ đạo và dẫn dắt sự phát triển của toàn bộ nhóm tôn giáo lớn đó.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mt

# Thiết lập cấu hình hiển thị biểu đồ
plt.rcParams.update({"font.family": "DejaVu Sans", "figure.dpi": 120,
                      "axes.spines.top": False, "axes.spines.right": False})

# 1. Khai phá cấu trúc nội bộ các nhánh của Cơ Đốc giáo trực tiếp từ df_clean
chrst_sub_cols = ['chrstprot', 'chrstcat', 'chrstorth', 'chrstang', 'chrstothr']
chrst_names = {
    'chrstprot': 'Tin Lành (Protestant)',
    'chrstcat': 'Công giáo (Catholic)',
    'chrstorth': 'Chính thống giáo (Orthodox)',
    'chrstang': 'Anh giáo (Anglican)',
    'chrstothr': 'Khác (Other)'
}

# Lọc các cột có thực
valid_chrst_subs = [c for c in chrst_sub_cols if c in df_clean.columns]

df_chrst = df_clean[['year'] + valid_chrst_subs].copy()
df_chrst['year'] = pd.to_numeric(df_chrst['year'], errors='coerce')

for col in valid_chrst_subs:
    if df_chrst[col].dtype == object:
        df_chrst[col] = df_chrst[col].astype(str).str.replace(',', '', regex=False).str.strip()
    df_chrst[col] = pd.to_numeric(df_chrst[col], errors='coerce').fillna(0)

# Tính tổng để đổi sang tỷ lệ % (100% chồng)
df_chrst['total'] = df_chrst[valid_chrst_subs].sum(axis=1)
df_chrst_pct = pd.DataFrame({'year': df_chrst['year']})
for col in valid_chrst_subs:
    df_chrst_pct[col] = (df_chrst[col] / df_chrst['total']) * 100
df_chrst_pct = df_chrst_pct.set_index('year')

# Vẽ biểu đồ Stacked Bar Chart cho Cơ Đốc giáo
fig, ax = plt.subplots(figsize=(11, 6))
colors_chrst = ['#2980b9', '#3498db', '#5dade2', '#85c1e9', '#aed6f1']
bottom = np.zeros(len(df_chrst_pct))

for i, col in enumerate(valid_chrst_subs):
    values = df_chrst_pct[col].values
    label = chrst_names.get(col, col)
    ax.bar(df_chrst_pct.index, values, bottom=bottom, label=label,
           color=colors_chrst[i % len(colors_chrst)], width=3.5, edgecolor='white', linewidth=0.5)
    bottom += values

# Thêm đầy đủ cấu hình định dạng để sửa lỗi incomplete input
ax.set_title("Cấu trúc phân nhánh bên trong của Cơ Đốc giáo (1945 - 2010)", fontsize=13, fontweight="bold", pad=15)
ax.set_xlabel("Năm", fontsize=11, fontweight="bold")
ax.set_ylabel("Tỷ trọng (%)", fontsize=11, fontweight="bold")
ax.set_ylim(0, 100)
ax.yaxis.set_major_formatter(mt.FuncFormatter(lambda v, _: f"{v:.0f}%"))
ax.xaxis.set_major_locator(mt.MultipleLocator(5))
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.legend(title="Nhánh Cơ Đốc giáo", bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()
plt.savefig("hinh_3_4_co_cau_co_doc_giao.png", bbox_inches="tight")
plt.show()
```
